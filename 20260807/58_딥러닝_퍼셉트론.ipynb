딥러닝(deep learning)

인간의 뇌는 뉴런이라는 세포로 이루어져 있고, 약 1,000억 개의 뉴런이 존재한다. 그리고 각 뉴런은 시냅스로 연결되는데, 자극의 크기가 특정값 이상이면 다음 뉴런으로 전달되고, 특정값 이하라면 다음 뉴런에 전달되지 않는다. 이때, 작극을 전단한는 기준인 특정값을 임계치라고 한다. 그래서 우리는 임계치를 넘긴 자극 신호를 전달받고 그에 해당하는 반응을 하게 되는 것이다.

이와 같은 생명 과학 분야의 아이디어를 머신러닝에 도입한 것을 인공 신경 망(Artifical Neural Network, ANN)이라고 하며 줄이면 신경망이라고 한다. 그리고 신경망을 기반으로 하는 학습하는 방법을 딥러닝이라고 부른다. 수많은 뉴런이 서로 연결되어 신호를 전달하는 것 처럼 딥러닝에서는 수많은 퍼셉트론이 연결되어 있어 연산 결과를 주고 받는다.

AND 연산이나 OR 연산의 경우 적합한 직선 하나를 생성해서 출려값 0과 1을 정확히 구분해 낼 수 있었다. 그러나 XOR 연산은 직선 하나로는 0과 1을 구분할 수 없다. 초기 딥러닝은 XOR 문제를 해결하는 방법으로 등장했고 신경망을 이용해서 기존 방법으로는 풀 수 없었던 XOR 문제를 풀 수 있었다. XOR 문제와 같은 간단한 문제를 해결하는 것으로 시작한 딥러닝은 점차 발전해 더욱 복잡한 문제를 푸는데 사용되었다.

퍼셉트론(perceptron)

퍼셉트론은 신경망의 최소 단위로 입력값과 출력값이 존재한다. n개의 입력을 합쳐서 벡터 $x$로 표현하고, n개의 가중치 값을 가중치 벡터 $w$라고 한다. 그리고 입력값 벡터와 가중치 벡터의 내적값은 출력값 $z$가 된다.

<img src="./perceptron.png" width="600" align="left" />

위의 그림에서 동그라미 노드를 의미하는데, 노드 뉴런에 해당된다. 입력 데이터에 가중치를 곱(내적)한 곱의 합을 가중 합(weighted sum)이라고 하고 이 가중 합은 출력값을 결정하는 활성화 함수(activation function)를 거치면서 최종 출력값을 반환한다.

활성화 함수에는 계단(step) 함수를 비롯해 시그모이드, 렐루, 리키 렐루 등이 있다. $b$는 편향(bias)으로 가중값에 더해지는 상수로 가중 합에 편향을 더함으로써 최종 출력값에 영향을 줄 수 있다.

퍼셉트론 실습

```python
# 입력 데이터
input_data = np.array([[2, 3], [5, 1]]) # 2차원 배열
print(input_data)

# 2차원 배열을 퍼셉트론에 입력하려면 1차원 형태로 변환시켜야 한다.
# ravel() 메소드는 배열을 무조전 1차원으로 변환한다.
x = input_data.ravel()
print(x)

# reshape() 메소드는 인수로 행과 열을 지정해서 원하는 차원으로 변환한다.
x = input_data.reshape(1, 4) # 2행 2열인 2차원 배열이 1행 4열인 2차원 배열로 변환된다.
print(x[0])
x = input_data.reshape(4) # 2행 2열인 2차원 배열이 배열 요소가 4개인 1차원 배열로 변환된다.
print(x)
x = input_data.reshape(-1) # 2행 2열인 2차원 배열이 배열 요소가 4개인 1차원 배열로 변환된다.
print(x)
```

```python
# 가중치 및 편향
w1 = np.array([2, 1, -3, 3]) # 가중치 벡터 w1
w2 = np.array([1, -3, 1, 3]) # 가중치 벡터 w2
print(w1, w2)

b1 = 3 # 편향 b1
b2 = 3 # 편향 b2
print(b1, b2)
```

```python
# 가중합
w = np.array([w1, w2]) # 가중치 벡터를 합쳐서 가중치 배열을 만든다.
print(w)
b = np.array([b1, b2]) # 편향을 합쳐서 편향 배열을 만든다.
print(b)
```

```python
# 내적
weigth_sum1 = np.dot(w, x) + b
weigth_sum2 = np.matmul(w, x) + b
weigth_sum3 = w @ x + b
print(weigth_sum1, weigth_sum2, weigth_sum3)
```

```python
# 출력층
# 시그모이드 함수를 활성화 함수로 사용해 가중합의 최종 점수를 확인한다.
result1 = 1 / (1 + np.exp(-weigth_sum1))
result2 = 1 / (1 + np.exp(-weigth_sum1))
result3 = 1 / (1 + np.exp(-weigth_sum1))
print(result1, result2, result3)
```
