인공 신경 망(Artifical Neural Network, ANN)

인공 신경망의 개념

다층 퍼셉트론(multi layer perceptron)은 이름 그대로 퍼셉트론의 층이 여러개라는 뜻으로 다층 퍼셉트론은 기존의 데이터 공간을 변형함으로써 기존 하나의 퍼셉트론으로 해결할 수 없었던 문제를 해결할 수 있게 되는 것이다. 이처럼 다수의 뉴런을 사용해서 만든 것은 인공 신경망이라고 한다.

<img src="./ann01.png" width="500" align="left" />

입력층(input layer)에 입력된 데이터는 출력층(output layer)에 도달하기 전에 은닉층(hidden layer)이라는 층을 거친 후 출력층에 도달한다. 은닉층은 하나의 층뿐만 아니라 다수의 층으로 정할 수 있고 딥러닝이라는 이름은 이 은닉층의 깊이가 깊다는 뜻(함수가 깊게 중첩된다.)에서 나온 이름이다. 입력층의 노드 개수는 피쳐 개수와 동일하고 출력층의 노드 개수는 분류하려는 클래스 개수와 같다.

출력층의 각 노드나 나타내는 숫자는 해당 클래스에 대한 점수(확률)이며 점수가 높을수록 해당 클래스에 속할 확률이 높다는 의미이고 최종적으로 점수가 가장 높은 클래스를 선정한다.

신경망 함수 $f(x) = f_3(f_2(f_1(x)))$와 같은 합성 함수 형태로 표현된다. $f_1$을 첫 번째 신경망 층, $f_1$를 두 번째 신경망 층, $f_3$을 세 번째 신경망 층이라고 생각하면 된다. 신경망 층이 깊을수록 함수의 개수는 많아지고 합성 함수도 복잡해진다. 이때 합성 함수의 개수는 모델의 깊이(depth)를 의미하는데, 딥러닝에서 deep이라는 용어가 바로 여기서 나온 용어이다.

오차 역전파(back propagation)

다층 퍼셉트론에서 최적의 값을 찾아가는 과정은 오차 역전파 방법을 사용한다.

<img src="./ann02.png" width="500" align="left" />

입력층 → 은닉층 → 출력층 순서대로 흘러가는 것을 순전파(forward propagation)라 하고 반대로 역전파는 출력층 → 은닉층 → 입력층 순서대로 반대로 거슬러 올라가는 방향이다. 오차 역전파를 기반으로 가중치와 편향값을 수정한 후 더 좋은 성능을 낼 수 있도록 모델을 개선한다.

활성화 함수(activation function)

활성화 함수는 딥러닝에서 입력값과 가중치 및 편향값을 계산해서 해당 노드를 활성화 할지를 결정하는 함수이다. 딥러닝에서는 결과값을 결정하는 여러가지의 활성화 함수가 존재한다.

① `항등 함수(identity function)`  
항등 함수 또는 선형 함수(linear function)라고도 부르는 이 함수는 입력과 출력이 같은 값을 가진다. 주로 회귀 문제에서의 활성화 함수로 사용된다.

항등 함수는 아래의 식을 따른다.

$$\phi(x) = x$$

② `계단 함수(step function)`  
계단 함수는 입력값이 0 이하일 경우 0을 출력하고, 0을 초과할 때만 1을 출력한다. 계단 함수의 출력은 0 또는 1로 오직 두 가지 값만 가진다는 특징이 있다.

계단 함수는 아래의 식을 따른다.

$$\phi(x) = 
\begin{cases}
0, \; x \leq 0 \\
1, \; x > 0
\end{cases}$$

계단 함수를 그림으로 나타내면 아래와 같다.

<img src="./ann03.png" width="400" align="left" />

입력값 $x$가 0을 기준으로 넘기 전과 후의 값에 따라 출려값이 바뀌는 것을 알 수 있다. 계단 함수는 사용하기 간단하다는 장점이 있지만 미분이 불가능하다는 단점이 있다.

③ `부호 함수(sign function)`  

부호 함수는 아래의 식을 따른다.

$$\phi(x) = 
\begin{cases}
-1, \; x < 0 \\
0, \; x = 0 \\
1, \; x > 0
\end{cases}$$

부호 함수를 그림으로 나타내면 아래와 같다.

<img src="./ann04.png" width="400" align="left" />

계단 함수와 비슷하게 생겼지만 0 또는 1만 출력하는 계단 함수와는 달리 -1, 0, 1 값을 출력하는 것을 알 수 있다.

④ `시그모이드 함수(sigmoid function)`

시그모이드 함수는 아래의 식을 따른다.

$$\phi(x) = \frac {1} {1 + e^{-x}}$$

시그모이드 함수를 그림으로 나타내면 아래와 같다.

<img src="./ann05.png" width="400" align="left" />

시그모이드 함수는 0과 1사이의 값을 출력한다.  
시그모이드 함수를 딥러닝에 적용했을 때 단점은 학습하는 과정에서 미분을 반복하면 기울기 값이 매우 작아지는 기울기 소실(vanishing gradiend) 문제가 발생할 수 있다는 것이다. 시그모이드 함수에서 $x$ 값이 지나치게 크거나 작을 경우, 미분값이 0에 가까워지고 이는 기울기 소실 문제를 발생시켜 학습 속도가 늘려질 수 있다.

⑤ `렐루 함수(Rectified Linear Function, ReLU)`

렐루 함수는 아래의 식을 따른다.

$$\phi(x) = 
\begin{cases}
0, \; x \leq 0 \\
x, \; x > 0
\end{cases}$$

위의 식은 아래와 같이 한 줄로도 표현할 수 있다.

$$\phi(x) = max(x, 0)$$

렐루 함수를 그림으로 나타내면 아래와 같다.

<img src="./ann07.png" width="500" align="left" />

렐루 함수는 $x$값이 0 이하라면 0을 출력하고, $x$값이 양수면 $x$값을 그대로 출력한다. 계단 함수, 부호 함수, 시그모이드 함수 함수와는 다르게 출력값의 상한선이 없다는 특징이 있다.

⑥ `리키 렐루 함수(Leaky ReLU Function)`

리키 렐루 함수는 아래의 식을 따른다.

$$\phi(x) = 
\begin{cases}
ax, \; x \leq 0 \\
x, \; x > 0
\end{cases}$$

위의 식은 $a \leq 1$ 이라면 아래와 같이 한 줄로도 표현할 수 있다.

$$\phi(x) = max(x, ax)$$

리키 렐루 함수를 그림으로 나타내면 아래와 같다.

⑦ `하이퍼볼릭 탄젠트 함수(Hyperbolic Tangent Function, tanh)`

하이퍼볼릭 탄젠트 함수는 아래의 식을 따른다.

$$\phi(x) = \frac {e^x - e^{-x}} {e^x + e^{-x}}$$

시그모이드 함수와의 관계를 통해 아래와 같이 표현할 수도 잇다.

$$\phi(x) = 2 \times sigmoid(2x) - 1$$

시그모이드 함수와 비슷하지만 시그모이드 함수의 출력 범위는 0부터 1사이인 반면에 하이퍼볼릭 탄젠트 함수의 출력 범위는 -1부터 1사이이다.
