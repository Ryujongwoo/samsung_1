데이터 전처리

우리가 실제로 접하게 되는 데이터는 사이킷런 라이브러리에서 제공하는 데이터와는 다르게 머신러닝 또는 딥러닝에 적용하기 전에 데이터를 가공하는 경우가 대부분이다. 이를 데이터 전처리 또는 피쳐 엔지니어링 등으로 부른다.

결측치(missing value) 처리

데이터셋의 일부에 데이터가 존재하지 않을 경우 결측치라 부르며, 머신러닝 또는 딥러닝에 적용하기 전에 결측치를 처리하는 것이 좋다.

```python
# 데이터프레임에 데이터가 존재하지 않는 결측치는 없지만 부적절한 값(이상치)이 존재한다.
# 나이에 1000이라는 값이 존재하는 것과 나머지 피쳐에 'unknown'이라는 값이 입력된 부적절한 값은 결측치로 처리하는 것이좋다.
df = pd.DataFrame([
    [42, 'male', 12, 'reading', 'class2'],
    [35, 'unknown', 3, 'cooking', 'class1'],
    [1000, 'female', 7, 'cycling', 'class3'],
    [1000, 'unknown', 21, 'unknown', 'unknown']
])

# 나이, 성별, 태어난 달, 취미, 레이블
df.columns = ['age', 'gender', 'month_birth', 'hobby', 'label']
df
```

```python
# pandas의 unique() 메소드로 각 시리즈의 유일한 값을 확인할 수 있다.
print('age: {}'.format(df.age.unique()))
print('gender: {}'.format(df.gender.unique()))
print('month_birth: {}'.format(df.month_birth.unique()))
print('hobby: {}'.format(df.hobby.unique()))
print('label: {}'.format(df.label.unique()))
```

```python
# 부적절한 데이터 값을 결측치로 바꾸기 위해서 np.nan를 사용한다.
df.loc[df.age > 150, ['age']] = np.nan
df.loc[df.gender == 'unknown', ['gender']] = np.nan
df.loc[df.month_birth > 12, ['month_birth']] = np.nan
df.loc[df.hobby == 'unknown', ['hobby']] = np.nan
df.loc[df.label == 'unknown', ['label']] = np.nan
df
```

```python
# 데이터셋의 각 열의 결측치 개수를 확인한다.
df.isnull().sum()
```

```python
df.info()
```

결측치를 처리하는 데 주로 사용되는 방법은 결측치를 삭제하거나 다른 값으로 변경하는 방법이 있다.

```python
# dropna() 메소드로 결측치를 삭제할 수 있다.
# axis 속성을 생략하거나 0으로 지정하면 결측치를 한 개라도 포함한 모든 행을 삭제하고 1로 지정하면 결측치가 한 개라도 포함한 모든 열을 삭제한다.
# inplace 속성을 True로 지정하면 실행 결과가 즉시 반영된다. 
df.dropna()
```

```python
df.dropna(axis=1)
```

```python
# how 속성을 'all'로 지정하면 모든 값이 결측치인 행만 제거한다.
df.dropna(how='all')
```

```python
# thresh 속성에 숫자를 지정하면 결측치를 제외한 값의 개수가 숫자보다 작은 행을 삭제한다.
df.dropna(thresh=1)
```

```python
# subset 속성에 열이름을 지정하면 지정한 열에 결츠치가 존재하는 행을 삭제한다.
df.dropna(subset=['gender'])
```

```python
# fillna() 메소드로 결측치를 대체한 값을 지정할 수 있다.
df.fillna(0)
```

```python
# value 속성으로 각 열의 결측치를 대체할 값이 저장된 딕셔너리를 지정해서 열별로 결측치를 대체할 값을 지정할 수 있다.
alter_values={'age': 0, 'gender': 'U', 'month_birth': 0, 'hobby': 'N', 'label': 'class4'}
df.fillna(value=alter_values, inplace=True)
```

기계 처리에 적합하도록 클래스 레이블 설정

label 열을 보면 문자열(str) 형태인데, 기계 처리에 적합하도록 하기 위해서 문자열을 상수(int)로 바꿔야 하는 경우가 있다.

사이킷런 라이브러리를 사용하지 않고 레이블 설정하기

```python
# 라벨링할 레이블을 얻어와서 오름차순으로 정렬한다.
label = list(df.label.values)
print(label)
label.sort() # 오름차순 정렬
# label.sort(reverse=True) # 내림차순 정렬
print(label)
```

```python
# 반복분을 실행해서 정렬된 레이블을 key로 index를 value로 하는 딕셔너리를 만든다.
label_dict = {}
for index, data in enumerate(label):
    print(index, data)
    label_dict[data] = index
label_dict
```

```python
# replace() 메소드로 딕셔너리에 저장된 데이터로 레이블을 라벨링한다.
df['label2'] = df.label.replace(label_dict)
df
```

사이킷런 라이브러리로 레이블 설정하기

```python
# 레이블 라벨링을 하기 위해 import 한다.
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder() # 라벨링을 실행할 객체를 만든다.
label = df.label.values
print(label)

# label_encoder.fit(label) # 학습시킨다. 라벨링한다.
# new_label = label_encoder.transform(label) # 학습(라벨링) 결과를 적용시킨다.
new_label = label_encoder.fit_transform(label) # 학습과 적용을 한 번에 실행한다.
print(new_label)

df['label3'] = new_label
df
```

데이터 스케일링(data scaling) - 정규화

피쳐는 제각기 다른 단위를 가지기 때문에 이로 인해 숫자 자체의 크기 차이가 발생한다. 데이터 스케일링은 데이터 값이 단위의 영향을 받지 않도록 변형하는 것을 의미한다. 전체 데이터를 학습 데이터와 테스트 데이터로 나누고 스케일링의 기준이 되는 값을 구할 때는 학습 데이터로만 구한다.

표준화(standard) 스케일링

대표적인 데이터 스케일링 방법으로 정규 분포(평균이 0이고 표준편차가 1)를 따르도록 변경하는 방법이다.

$$\frac{x_i-\bar x}{\sigma}$$

위 식의 의미는 기존 데이터 값에서 평균을 뺀 후 표준편차로 나눈다는 의미이다.

```python
# 표준화 스케일링을 사용하기 위해 import 한다.
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler() # 표준화 스케일링 객체를 만든다.
data = df[['month_birth']]
print(data)

# scaler.fit(data) # 학습시킨다. 표준화 스케일러를 실행한다.
# new_scaler = scaler.transform(data) # 학습(표준화 스케일러) 결과를 적용시킨다.
new_scaler = scaler.fit_transform(data) # 학습과 적용을 한 번에 실행한다.
print(type(new_scaler))
print(new_scaler)
```

```python
# 실제로 표준화 스케일링 적용 후 데이터 평균이 0이고 표준편차가 1인지 확인한다.
print('평균: {:.1f}, 표준편차: {:.1f}'.format(np.mean(new_scaler), new_scaler.std()))
```

로버스트(robust) 스케일링

표준화 스케일링을 변형한 방법으로 중앙값(median)과 사분위수(quantile)를 사용한다. 극단값의 영향을 거의 받지 않는다는 장점이 있다.

$$\frac{x_i -q_2}{q_3-q_1}$$

위 식에서 $q_1$은 1사분위수, $q_2$은 2사분위수(중위수), $q_3$은 3사분위수를 의미한다.

```python
# 로버스트 스케일링을 사용하기 위해 import 한다.
from sklearn.preprocessing import RobustScaler
```

```python
scaler = RobustScaler() # 로버스트 스케일링 객체를 만든다.
data = df[['month_birth']]
print(data)

# scaler.fit(data) # 학습시킨다. 로버스트 스케일러를 실행한다.
# new_scaler = scaler.transform(data) # 학습(로버스트 스케일러) 결과를 적용시킨다.
new_scaler = scaler.fit_transform(data) # 학습과 적용을 한 번에 실행한다.
print(type(new_scaler))
print(new_scaler)
```

최소-최대(min-max) 스케일링

데이터 값의 범위를 최소값(0)과 최대값(1) 사이의 값으로 범위를 계산한다.

$$\frac{x_i-min(x)}{max(x)-min(x)}$$

```python
# 최소-최대 스케일링을 사용하기 위해 import 한다.
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler() # 최소-최대 스케일링 객체를 만든다.
data = df[['month_birth']]
print(data)

# scaler.fit(data) # 학습시킨다. 최소-최대 스케일러를 실행한다.
# new_scaler = scaler.transform(data) # 학습(최소-최대 스케일러) 결과를 적용시킨다.
new_scaler = scaler.fit_transform(data) # 학습과 적용을 한 번에 실행한다.
print(type(new_scaler))
print(new_scaler)
```

노멀(normalizer) 스케일링

벡터의 유클리드 거리(두 점 사이의 거리 계산에 사용)가 1이 되로록 데이터 값을 변경한다. 주로 벡터의 길이는 상관없고 방향(각도)만 고려할 때 사용한다.  
표준화, 로버스트, 최소-최대 스케일링은 열 기준인 것과 다르게 노멀 스케일링은 행 기준이다.

$$\frac{x_i}{\sqrt{x_i^2+y_i^2+z_i^2}}$$

```python
# 노멀 스케일링을 사용하기 위해 import 한다.
from sklearn.preprocessing import Normalizer
```

```python
scaler = Normalizer() # 노멀 스케일링 객체를 만든다.
data = df[['age', 'month_birth']]
print(data)

# scaler.fit(data) # 학습시킨다. 노멀 스케일러를 실행한다.
# new_scaler = scaler.transform(data) # 학습(노멀 스케일러) 결과를 적용시킨다.
new_scaler = scaler.fit_transform(data) # 학습과 적용을 한 번에 실행한다.
print(type(new_scaler))
print(new_scaler)
```

데이터 스케일링 과정에서 fit() 메소드는 학습 데이터에 대해서만 사용하며, 테스트 데이터에는 fit() 메소드를 사용하지 않고 transform() 메소드만 사용한다.  
이유는 스케일러는 학습 데이터를 기반으로 만들어져야 하기 때문이다. 만약에 학습 데이터와 테스트 데이터에 모두 fit() 메소드를 실행해서 각각의 데이터로 스케일링한다면 학습 데이터와 테스트 데이터의 분포가 다르기때문에 학습 데이터와 테스트 데이터의 스케일링 범위 및 파라미터가 달라진다.
