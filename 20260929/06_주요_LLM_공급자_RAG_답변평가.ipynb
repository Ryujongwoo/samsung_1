## JSONL 파일 읽기

JSONL(JSON Line) 형식의 데이터를 읽어서, LangChain의 표준으로 사용하는 Document 객체 형태로 변환한다.

LangChain의 가장 기본적인 데이터 단위인 본문(page_content)과 부가 정보(metadata)를 묶어서 관리하는 Document를 import 한다.  
`from langchain_core.documents import Document`  
JSON 또는 JSONL 형식의 문자열을 파이썬에서 처리할 수 있도록 리스트나 딕셔너리 형태로 변환하거나 그 반대 작업을 하기위해 import 한다.  
`import json`

```python
final_docs = []

# JSONL 파일을 읽기 위해서 final_docs.jsonl 파일을 읽어들인다.
# './data' 폴더의 'final_docs.jsonl' 파일을 바이너리 형태의 입력용으로 open 한다.
# 파일 모드는 인코딩 문제를 방지하기 위해 'rb'는 Read Binary의 약자인 '읽기 전용' 및 '바이너리 모드'로 연다.
# with 구문을 사용했으므로 with 블록의 모든 작업이 완료되면 파일이 자동으로 안전하게 닫힌다. 입력용 파일을 닫는 이유는 메모리를 효율적으로 사용하기 위해서이다.
with open('./data/final_docs.jsonl', 'rb') as file:
    for line in file:
        # json 라이브러리의 loads() 메소드는 인수로 지정된 문자열을 리스트나 딕셔너리 형태로 변환한다.
        item = json.loads(line)
        # 딕셔너리에서 'page_content'와 'metadata'를 꺼내서 Document 객체를 만들어 final_docs 리스트에 추가한다.
        doc = Document(page_content=item['page_content'], metadata=item['metadata'])
        final_docs.append(doc)
# ===== with

print(len(final_docs))
final_docs
```

## 벡터검색기 - 시맨틱 검색, 맥락의 의미

임베딩 기술을 사용하여 텍스트의 의미를 숫자로 변환하고, 이를 바탕으로 질문과 가장 유사한 답변을 찾아내는 벡터검색기를 정의한다.

오픈소스 벡터 데이터이터베이스로 텍스트를 숫자로 바꾼 데이터를 저장하고 빠르게 검색하기 위해 Chroma를 import 한다.  
`from langchain_chroma import Chroma`  
오픈소스 모델 허브인 Hugging Face에서 제공하는 임베딩 모델을 사용하기 위해서 HuggingFaceEmbeddings를 import 한다.  
`from langchain_huggingface.embeddings import HuggingFaceEmbeddings`

벡터저장소에 사용한 임베딩 모델을 설정한다.

```python
# HuggingFaceEmbeddings 클래스의 생성자로 임베딩 모델 이름을 넘겨서 임베딩 모델을 설정한다.
# 임베딩이란 컴퓨터 문장을 이해할 수 있는 숫자들로 바꾸는 과정이다.
embeddings_model = HuggingFaceEmbeddings(model='BAAI/bge-m3')
```

벡터저장소를 로드한다.

```python
# Chroma 클래스의 생성자로 임베딩 모델, 테이블 이름, 저장된 폴더를 넘겨서 벡터스토어를 로드한다.
chroma_db = Chroma(
    collection_name='hf_bge_m3',
    embedding_function=embeddings_model,
    persist_directory='./chroma_db',
)
```

벡터검색기를 만든다.

```python
# as_retriever() 메소드로 질문과 유사도를 계산해서 의미적으로 가까운 가져올 문서 개수를 넘겨서 벡터검색기를 만든다.
chroma_k_retriever = chroma_db.as_retriever(search_kwargs={'k': 2})
```

벡터검색기를 실행한다.

```python
query = '테슬라의 회장은 누구인가요?'
# invoke() 메소드로 쿼리를 넘겨서 거리가 가장 가까운 문서를 찾아 반환한다.
retriever_docs = chroma_k_retriever.invoke(query)
```

```python
print(f'쿼리: {query}')
print('검색 결과')
for doc in retriever_docs:
    print(doc.page_content)
    print(doc.metadata)
    print('-' * 100)
```

## BM25검색기 - 키워드 기반, 단어의 일치

한국어 형태소 분석기인 Kiwi를 활용하여, 텍스트의 키워드가 얼마나 일치하는지를 기준으로 문서를 찾아주는 KiWiBM25 검색기를 정의한다.

한국어 문장을 형태소 단위로 나눠주는 도구인 KiwiTokenizer를 import 한다.  
한국어 형태소 분석기인 Kiwi를 활용해서 관련 문서를 찾아주는 KiWiBM25RetrieverWithScore를 import 한다.

```python
from krag.tokenizers import KiwiTokenizer
from krag.retrievers import KiWiBM25RetrieverWithScore
```

한국어를 분석에 사용할 토크나이저를 설정한다.

```python
# KiwiTokenizer 클래스의 생성자로 토크나이저 모델과 오타 교정 여부를 넘겨서 토크나이저를 설정한다.
kiwi_tokenizer = KiwiTokenizer(
    # krag에서 가장 권장되는 모델인 특정 단어 뒤에 어떤 형태소가 나올 확률이 높은지 계산해서 문맥을 파악하는 knlm을 지정한다.
    model_type='knlm',
    # 형태소 분석시 일반적인 수준의 오타를 처리하는 'basic'을 지정한다.
    typos='basic'
)
```

KiWiBM25 검색기를 만든다.

```python
# KiWiBM25RetrieverWithScore 클래스의 생성자로 토큰화할 문서, kiwi 토크나이저, 가져올 문서 개수, 검색 점수 임계값을 넘겨서 KiWiBM25 검색기를 만든다.
kiwibm25_k_retriever = KiWiBM25RetrieverWithScore(
    # KiWiBM25 검색기를 만들기 위해서 토큰화 할 문서 리스트를 지정한다.
    documents=final_docs,
    # 형태소 분석에 사용할 Kiwi 토크나이저를 지정한다.
    kiwi_tokenizer=kiwi_tokenizer,
    # 검색해서 가장 관련성이 높은 점수 순서로 가져올 문서의 개수를 지정한다.
    k=2,
    # 검색 점수의 임계값을 지정한다. 0.0으로 지정하면 점수에 상관없이 모든 결과를 가져온다.
    threshold=0.0
)
```

KiWiBM25 검색기를 실행한다.

```python
query = '테슬라의 회장은 누구인가요?'
retriever_docs = kiwibm25_k_retriever.invoke(query)
```

```python
print(f'쿼리: {query}')
print('검색 결과')
for doc in retriever_docs:
    print(doc.page_content)
    print(doc.metadata)
    print('-' * 100)
```

## Ensemble 검색기 - Hybrid Search

서로 다른 2개 이상의 검색기(Chroma(의미 기반), KiwiBM25(키워드 기반))를 하나로 합쳐서 각각의 장점만 취하는 Ensemble 검색기를 정의한다.

여러 개의 검색 알고리즘을 결합하여 더 나은 검색 결과를 만들어내는 EnsembleRetriever를 import 한다.

```python
from langchain.retrievers import EnsembleRetriever
```

Ensemble 검색기를 만든다.

```python
# Chroma 벡터저장소나 KiwiBM25 저장소를 새로 만들지 않고 기존 저장소의 검색기의 상위 k을 아래와 같이 변경할 수 있다.
chroma_k_retriever.search_kwargs['k'] = 4
kiwibm25_k_retriever.k = 4
```

```python
# EnsembleRetriever 클래스의 생성자로 사용할 서로 다른 검색기 목록과 각 검색기의 비중을 넘겨서 Ensemble 검색기를 만든다.
ensemble_k_retriever = EnsembleRetriever(
    # 사용할 서로 다른 검색기 목록을 리스트 형태로 지정한다.
    retrievers=[chroma_k_retriever, kiwibm25_k_retriever],
    # 각 검색기별 가중치를 리스트 형태로 지정한다. 이 수치를 조절해서 검색 성능을 튜닝할 수 있다.
    weights=[0.5, 0.5]
)
```

Ensemble 검색기를 실행한다.

```python
query = '테슬라의 회장은 누구인가요?'
retriever_docs = ensemble_k_retriever.invoke(query)
```

```python
print(f'쿼리: {query}')
print('검색 결과')
for doc in retriever_docs:
    print(doc.page_content)
    print(doc.metadata)
    print('-' * 100)
```

## CrossEncoder 알고리즘에 기반하여 순위를 다시 매긴다. - Rerank

Hugging Face에서 제공하는 CrossEncoder 모델을 불러와서 재순위화 하기 위해서 HuggingFaceCrossEncoder를 import 한다.  
LangChain에서 RAG 시스템의 성능 향상시키기 위해 사용하는 문서 재순위화 하기 위해서 CrossEncoderReranker import 한다.  
LangChain에서 기본 검색기에서 가져온 문서 중 질문과 진짜 관련된 내용만 압축, 선별해서 LLM에 전달하기 위해서 ContextualCompressionRetriever import 한다.

```python
from langchain_community.cross_encoders import HuggingFaceCrossEncoder
from langchain.retrievers.document_compressors import CrossEncoderReranker
from langchain.retrievers import ContextualCompressionRetriever
```

재순위화에 사용할 재순위화 모델을 불러온다.

```python
# HuggingFaceCrossEncoder 클래스의 생성자로 재순위화에 사용할 모델 이름을 넘겨서 재순위화 모델을 설정한다. 
model = HuggingFaceCrossEncoder(model_name='BAAI/bge-reranker-v2-m3')
```

불러온 재순위화 모델을 이용해서 실제로 재순위화 작업을 하는 모델을 만든다.

```python
# CrossEncoderReranker 클래스의 생성자로 재순위화에 사용할 모델과 점수에 따라 가져올 문서 개수를 넘겨서 질문과 관련된 내용만 압축, 선별하는 모델을 만든다.
re_ranker = CrossEncoderReranker(model=model, top_n=3)
```

기본 검색기와 재순위화 검색기를 이용하는 최종 검색기를 만든다.

```python
# ContextualCompressionRetriever 클래스의 생성자로 기본 검색기와 기본 압축기를 넘겨서 최종 검색기를 만든다.
cross_encoder_reranker_retriever = ContextualCompressionRetriever(
    # 기본 검색기를 Chroma(의미 기반), KiwiBM25(키워드 기반)를 하나로 합쳐서 각각의 장점만 취하는 Ensemble 검색기를 지정한다.
    base_retriever=ensemble_k_retriever,
    # 기본 검색기에서 얻어온 문서를 재순위화를 하는 모델을 지정한다.
    base_compressor=re_ranker
)
```

ensemble_k_retriever 검색기에서 얻어온 문서의 재순위화를 실행한다.

```python
query = '테슬라의 회장은 누구인가요?'
retriever_docs = cross_encoder_reranker_retriever.invoke(query)
```

```python
print(f'쿼리: {query}')
print('검색 결과')
for doc in retriever_docs:
    print(doc.page_content)
    print(doc.metadata)
    print('-' * 100)
```

OpenAI의 GPT 모델은 현재 가장 널리 알려진 LLM 중 하나로 GPT-3.5, GPT-4, GPT-5 등 다양한 버전을 제공한다.

`05_고급검색기법` 파일의 `RAG 파이프라인으로 적용하기` 부분을 함수 형태로 만들어 실행한다.

LLM에게 전달할 메시지를 정의하기 위해 ChatPromptTemplate를 import 한다.  
`from langchain_core.prompts import ChatPromptTemplate`  
LangChain에서 제공하는 OpenAI의 대화형 AI을 활용하기 위해 ChatOpenAI를 import 한다.  
`from langchain_openai import ChatOpenAI`  
LLM이 응답한 객체에서 문자열만 뽑아내기 위해 StrOutputParser를 import 한다.  
`from langchain_core.output_parsers import StrOutputParser`

```python
# 여기서는 질문을 가공 없이 그대로 프롬프트에 넘기기 위해서 RunnablePassthrough를 import 한다.
from langchain_core.runnables import RunnablePassthrough
```

검색기와 LLM 모델을 인수로 넘겨받아 프롬프트를 만들고 LLM 모델에 전달해서 LLM 모델이 응답한 결과를 문자열만 얻어오는 RAG chain을 리턴하는 함수를 정의한다.

```python
def create_rag_chain(retriever, llm):
    # 프롬프트 템플릿
    template = '''
    당신은 유능한 나만의 인공지능 비서입니다.
    당신이 수행할 과제는 제공된 문맥(context)을 바탕으로 다음 질문에 답변하는 것입니다.
    다음 문맥(context)에만 근거하여 질문에 답변하고 외부 정보나 지식은 절대로 사용하지 마세요.
    만약 문맥(context)이 질문과 관련이 없다면, '답변에 필요한 근거를 찾지 못했습니다.'라고 답변하세요.

    [문맥(context)]
    {context}

    [질문]
    {question}

    [답변]
    '''
    prompt = ChatPromptTemplate.from_template(template=template)
    
    # 검색기가 검색한 문서들에서 본문 부분만 꺼내서 하나의 긴 문자열로 연결하는 함수
    def format_docs(docs):
        return '\n\n'.join([doc.page_content for doc in docs])
    
    # RAG Chain을 만들어서 리턴시킨다.
    return {'context': retriever | format_docs, 'question': RunnablePassthrough()} | prompt | llm | StrOutputParser()
```

```python
# LLM 모델 정의
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0.0)

# 검색기와 LLM 모델을 넘겨서 RAG chain을 얻어오는 함수 실행
openai_rag_chain = create_rag_chain(cross_encoder_reranker_retriever, llm)

# RAG chain 실행
question = '테슬라의 회장은 누구인가요?'
answer = openai_rag_chain.invoke(question)
```

```python
print(f'질문: {question}')
print(f'답변: {answer}')
```

# RAG 답변 평가(정량적 지표) - Embedding Distance

임베딩 거리는 두 문장의 의미적 유사성을 벡터 공간에서 측정하는 방법으로 임베딩 거리를 사용하여 예측 문자열과 참조 레이블 문자열 간의 의미적 유사성을 측정하고 계산된 `거리 점수가 낮을수록` 두 문자열의 의미가 더 `유사함`을 나타낸다. 이 방법은 단순 문자열 비교보다 더 풍부한 의미적 평가가 가능하다.

작동 원리  
① 예측 문자열과 참조 레이블 문자열을 벡터로 임베딩한다.  
② 두 벡터 간의 거리(코사인 유사도나 유클리디안 거리)를 계산한다.  
③ 거리 점수가 낮을수록 의미적으로 더 유사함을 의미한다.

장점  
단순 문자열 비교보다 더 깊은 의미 관계를 포착하고 한 번 임베딩을 생성하면 빠르게 비교한다.  

단점  
전체 문맥을 완전치 캡처하지 못할 수 있다.

`from langchain_openai import ChatOpenAI`

LangChain의 다양한 평가 도구(Evaluator)를 생성하고 불러오기 위해 load_evaluator를 import 한다.  
사용할 평가 방식의 종류를 정의해 놓은 목록(enum)을 불러오기 위해 EvaluatorType를 import 한다.  
두 문장 사이의 `거리`를 계산하는 수학적 알고리즘을 불러오기 위해 EmbeddingDistance를 import 한다.  

```python
from langchain.evaluation import load_evaluator, EvaluatorType, EmbeddingDistance
```

평가 도구(Evaluator)를 만든다.

load_evaluator() 함수로 문장 유사도 비교, 정답 여부 판단 등 다양한 목적의 평가 도구를 만들 수 있다.

## EvaluatorType은 LangChain의 평가 도구를 지정한다.

`평가자                    | 사용 시점                           | 핵심 특징                          `  
`STRING_DISTANCE           | 글자 표기의 일치도가 중요할 때      | 빠르고 비용 없음, 단순 텍스트 비교 `  
`EMBEDDING_DISTANCE        | 표현은 달라도 뜻이 같은지 볼 때     | 임베딩 모델 사용, 문맥적 유사도    `  
`JSON_EDIT_DISTANCE        | JSON schema 출력을 검증할 때        | 스키마 구조 및 값의 차이 측정      `  
`QA                        | RAG 답변의 사실 여부를 검증할 때    | 입력 컨텍스트 기반 정확도 측정     `  
`CRITERIA                  | 정답 없이 유용성/안전성 등을 볼 때  | LLM 판단, 개별 출력 평가           `  
`LABELED_CRITERIA          | 정답(Reference) 기준으로 평가할 때  | LLM 판단, 정답지 대조              `  
`PAIRWISE_CRITERIA         | A/B 테스트로 더 나은 모델을 고를 때 | 두 출력 간 비교(No Reference)      `  
`LABELED_PAIRWISE_CRITERIA | 정답지를 두고 A/B 테스트를 할 때    | 두 출력 간 비교(With Reference)    `

## EMBEDDING_DISTANCE를 사용할 때, 내부적으로 거리를 계산하는 방식은 EmbeddingDistance를 사용해서 지정한다.

`옵션      | 측정 기준       | 주요 특징               | 추천 사용 시점                                                 `  
`COSINE    | 각도 차이       | 크기 무시               | 방향성 집중, 일반적인 텍스트/문장 임베딩 유사도 검색(기본 권장)`  
`EUCLIDEAN | 직선거리(L2)    | 공간상 절대 거리        | 정규화된 벡터 비교, 일반 다차원 클러스터링                     `  
`MANHATTAN | 축 이동거리(L1) | 이상치에 비교적 덜 민감 | 희소(Sparse) 데이터, 고차원 특성 공간                          `  
`CHEBYSHEV | 최대 차이값     | 단일 최대 오차 측정     | 차원별 최대 허용 오차가 중요한 특수 평가                       `  
`HAMMING   | 불일치 개수     | 비트/이진 연산 최적화   | 양자화(Binary) 임베딩의 초고속 유사도 비교                     `

```python
# load_evaluator() 함수의 인수로 평가 방식, EMBEDDING_DISTANCE의 거리를 재는 방식, LLM 모델을 넘겨서 평가 도구를 만든다.
embedding_evaluator = load_evaluator(
    # EvaluatorType을 이용해서 평가 방식을 지정한다.
    evaluator=EvaluatorType.EMBEDDING_DISTANCE,
    # EmbeddingDistance를 이용해서 EMBEDDING_DISTANCE의 거리를 재는 방식을 지정한다.
    # 텍스트 분석에 가장 많이 사용되는 방식으로 결과값이 0에 가까울수록 의미가 거의 일치하며, 1에 가까울수록 의미가 멀다는 뜻이다.
    distance_metric=EmbeddingDistance.COSINE,
    # 텍스트를 임베딩하거나 처리할 때 사용할 LLM 모델을 지정한다.
    llm=ChatOpenAI(model='gpt-4o-mini', temperature=0.0)
)
```

의미가 대조적인 문장 테스트

```python
# evaluate_strings() 메소드의 인수로 예측값과 참조값을 넘겨서 문장을 비교한다.
result1 = embedding_evaluator.evaluate_strings(
    # 예측값, AI가 생성했거나 테스트하고 싶은 문장
    prediction='나는 학교에 갈 것이다.',
    # 참조값, 예측값과 비교할 정답 문장
    reference='나는 집에 있을 것이다.'
)
print(result1)
```

'학교'와 '집', '가다'와 '있다'는 의미적으로 거리가 멀어서 상대적으로 높은 수치가 나온다.

질문과 정답을 인수로 넘겨받아 위에서 따로따로 작업한 5개의 모델의 임베딩 거리를 계산해서 리턴하는 함수를 만든다.

```python
def embedding_distance_all_model(question, answer):
    # 모델별 임베딩 거리 평가 결과를 저장할 빈 딕셔너리를 만든다.
    results = {}
    # 사용할 모델 이름과 이미 생성해둔 각 모델의 RAG chain을 쌍으로 묶어서 반복하며 Embedding Distance를 계산한다.
    rag_chain_dict = {
        'OpenAI': openai_rag_chain, 'Claude': anthropic_rag_chain, 'Gemini': google_genai_rag_chain, 'Ollama': ollama_rag_chain, 'Groq': groq_rag_chain
    }
    # print(rag_chain_dict)
    for model_name, rag_chain in rag_chain_dict.items():
        # print(model_name, rag_chain)
        # 각 모델의 RAG chain에 질문을 넘겨서 얻어온 답변(예측값)을 원래 정답과 임베딩 거리를 계산한다.
        prediction = rag_chain.invoke(question)
        distance_score = embedding_evaluator.evaluate_strings(prediction=prediction, reference=answer)
        # 모델별 평가 결과를 저장할 딕셔너리에 모델 이름을 Key로 각 모델의 유사도 점수를 Value로 넣어준다.
        results[model_name] = f'{distance_score["score"]:.5f}'
    return results
```

```python
embedding_distance_all_model(question, answer)
```

질문과 정답이 저장된 데이터프레임을 인수로 넘겨받아 위에서 따로따로 작업한 5개의 모델의 임베딩 거리를 계산해서 데이터프레임으로 리턴하는 함수를 만든다.

```python
def embedding_distance_qa_test(df_qa_test):
    # print(len(df_qa_test))
    # 데이터프레임으로 만들 모델별 평가 점수를 저장할 빈 리스트를 만든다.
    results = []
    # 인수로 넘겨받은 데이터 건 수 만큼 반복하며 모델별 평가 점수를 계산한다.
    for i in range(len(df_qa_test)):
        # 인수로 넘겨받은 데이터프레임의 i번째 행에서 질문과 답변을 얻어온다.
        question = df_qa_test.iloc[i]['question']
        answer = df_qa_test.iloc[i].answer
        # print(question, answer)
        # 질문과 정답을 인수로 넘겨받아 5개의 모델의 임베딩 거리를 계산해서 리턴하는 함수를 실행한다.
        result = embedding_distance_all_model(question, answer)
        # 모델별 평가 점수가 저장된 딕셔너리를 모델별 임베딩 거리를 저장할 리스트에 추가한다.
        results.append(result)
    # 모델별 평가 점수가 저장된 리스트를 데이터프레임으로 만들어 리턴시킨다.
    return pd.DataFrame(data=results)
```

# RAG 답변 평가(정량적 지표) - Cross Encoder

임베딩 거리는 개별 문장을 따로따로 인코딩하는 방식이고 Cross Encoder는 두 문장을 동시에 인코딩하여 직접적인 유사성을 평가한다.  
두 문장의 상호 작용을 직접 모델링하므로, 일반적으로 더 정확한 유사성 점수를 제공하고 의미적 유사성을 더 정확하게 측정할 수 있지만, 계산 비용이 더 높다.

작동 원리  
① 두 문장을 하나의 입력으로 모델에 제공한다.  
② 모델이 두 문장의 관계를 직접 분석하여 유사성 점수를 출력한다.  

장점  
두 문장의 관계를 직접 모델링하여 더 정확한 유사성 평가가 가능하므로 전체 문맥을 고려한 평가가 가능하다. 

단점  
각 문장 쌍마다 별도의 모델 실행이 필요하여 처리 시간이 길고 계산 비용이 높다.

일반적인 Bi-Encoder(문장을 각각 벡터로 만드는 방식)와 달리, 두 문장을 동시에 모델에 입력하여 관계를 직접 계산하는 Cross Encoder를 사용하기 위해 import 한다.

```python
from sentence_transformers import CrossEncoder
```

CrossEncoder를 활용해서 두 문장(질문과 답변 후보) 간의 연관성 점수를 계산하는 함수를 만든다.  
실제 정답과 LLM이 생성한 답변, CrossEncoder에 사용할 모델을 인수로 넘겨받아 위에서 따로따로 작업한 5개의 모델의 평가 점수를 계산해서 리턴하는 함수를 만든다.

```python
# 매번 모델을 새로 로드하므로 속도가 느려므로 모델을 로드를 함수 외부에서 실행한다.
cross_encoder_model = CrossEncoder(model_name_or_path='BAAI/bge-reranker-v2-m3')
```

```python
def cross_encoder_score(answer: str, prediction: str, model_name: str = 'BAAI/bge-reranker-v2-m3') -> float:
    # CrossEncoder 클래스의 생성자로 Cross Encoder에 사용할 모델을 넘겨서 CrossEncoder 모델을 만든다.
    # 함수가 호출될 때마다 이 라인이 실행되면 매번 모델을 새로 로드하므로 속도가 느려질 수 있다. 함수 밖에서 모델을 한 번만 선언하는 것이 효율적이다.
    # cross_encoder_model = CrossEncoder(model_name_or_path=model_name)
    # 입력 데이터를 준비한다.
    # CrossEncoder는 두 문장을 쌍으로 입력받아야 하므로 여러 개의 쌍을 한꺼번에 처리할 수 있도록 리스트 안에 리스트가 있는 [[문장1, 문장2]] 형태로 구성한다.
    sentence_pairs = [[answer, prediction]]
    # predict() 메소드에 준비된 입력 데이터를 모델에 넘겨서 유사도를 예측한다.
    score = cross_encoder_model.predict(sentence_pairs)
    # predict() 메소드 실행 결과는 넘파이 배열 형태이므로 서식 적용이 안되므로 넘파이 배열의 0번째 인덱스 값만 리턴시킨다.
    return score[0]
```

Embedding Distance는 계산된 거리 점수가 낮을수록 두 문자열의 의미가 더 유사함을 의미하지만 Cross Encoder는 계산된 유사도 점수가 높을수록 두 문장의 의미적 연관성이 높다는 의미이다.

```python
print('질문:', question) # 사용자가 던진 원래 질문
print('답변:', answer) # 실제 정답
print('OpenAI 모델 예측:', openai_prediction) # LLM이 생성한 답변

score = cross_encoder_score(answer, openai_prediction)
print(f'유사도 점수: {score:.5f}')
```

질문과 정답을 인수로 넘겨받아 위에서 따로따로 작업한 5개의 모델의 유사도 점수를 계산해서 리턴하는 함수를 만든다.

```python
def cross_encoder_all_model(question: str, answer: str) -> dict:
    # 모델별 유사도 점수 평가 결과를 저장할 빈 딕셔너리를 만든다.
    results = {}
    # 사용할 모델 이름과 이미 생성해둔 각 모델의 RAG chain을 쌍으로 묶어서 반복하며 Embedding Distance를 계산한다.
    rag_chain_dict = {
        'OpenAI': openai_rag_chain, 'Claude': anthropic_rag_chain, 'Gemini': google_genai_rag_chain, 'Ollama': ollama_rag_chain, 'Groq': groq_rag_chain
    }
    for model_name, rag_chain in rag_chain_dict.items():
        # 각 모델의 RAG chain에 질문을 넘겨서 얻어온 답변(예측값)을 얻어온다.
        prediction = rag_chain.invoke(question)
        print(model_name)
        print(prediction)
        print('-' * 100)
        # 각 모델의 RAG chain에 질문을 넘겨서 얻어온 답변(예측값)과 원래 정답과 유사도 점수를 계산한다.
        similarity_score = cross_encoder_score(answer, prediction)
        # print(similarity_score)
        # 모델별 평가 결과를 저장할 딕셔너리에 모델 이름을 Key로 각 모델의 유사도 점수를 Value로 넣어준다.
        results[model_name] = f'{similarity_score:.5f}'
    return results
```

```python
cross_encoder_all_model(question, answer)
```

질문과 정답이 저장된 데이터프레임을 인수로 넘겨받아 위에서 따로따로 작업한 5개의 모델의 유사도 점수를 계산해서 데이터프레임으로 리턴하는 함수를 만든다.

```python
def cross_encoder_qa_test(df_qa_test):
    # 데이터프레임으로 만들 모델별 평가 점수를 저장할 빈 리스트를 만든다.
    results = []
    # 인수로 넘겨받은 데이터 건 수 만큼 반복하며 모델별 평가 점수를 계산한다.
    for i in range(len(df_qa_test)):
        # 인수로 넘겨받은 데이터프레임의 i번째 행에서 질문과 답변을 얻어온다.
        question = df_qa_test.iloc[i]['question']
        answer = df_qa_test.iloc[i].answer
        # 질문과 정답을 인수로 넘겨받아 5개의 모델의 유사도 점수를 계산해서 리턴하는 함수를 실행한다.
        result = cross_encoder_all_model(question, answer)
        # 모델별 평가 점수가 저장된 딕셔너리를 모델별 유사도 점수를 저장할 리스트에 추가한다.
        results.append(result)
    # 모델별 평가 점수가 저장된 리스트를 데이터프레임으로 만들어 리턴시킨다.
    return pd.DataFrame(data=results)
```

# RAG 답변 평가(정량적 지표) - ROUGE Metric

ROUGE(Recall-Oriented Understudy for Gisting Evaluation)  
ROUGE Metric은 두 문장의 유사도를 평가하는데 널리 사용되는 방법으로, 특히 텍스트 요약과 기계 번역 분야에서 유용하다.  
단어 중첩을 기반으로 하여 계산이 빠르고 해석이 쉽다는 장점이 있지만, 깊은 의미적 유사성을 포착하는데 한계가 있고 사용 목적과 컨텍스트에 따라 적절히 선택하거나 다른 Metric과 조합해서 사용한다.

rougeN: N에는 1, 2와 같은 숫자를 입력해서 N-gram 중첩을 측정한다.  
rougeL: 최장 공통 부분 수열을 기반으로 유사성을 측정한다.  

장점  
단순한 단어 중접 기반으로 빠르게 계산할 수 있고 결과를 쉽게 이해하고 해석이 가능하다. 텍스트 요약 분야에서 널리 사용된다.  

단점  
단어 순서 변경에 민감할 수 있고 깊은 의미적 관계를 포착하는데 한계가 있다.

한국어 텍스트 비교에 특화된 korouge_score 라이브러리를 사용해서 모델의 답변이 정답과 얼마나 겹치는지 ROUGE 지표를 계산하기 위해 rouge_scorer를 import 한다.

```python
from korouge_score import rouge_scorer
```

rouge_scorer를 활용해서 모델의 답변과 정답이 얼마나 겹치는지 ROUGE 지표 점수를 계산하는 함수를 만든다.  
실제 정답과 LLM이 생성한 답변, 계산할 ROUGE 지표 종류를 인수로 넘겨받아 위에서 따로따로 작업한 5개의 모델의 ROUGE 지표 점수를 계산해서 리턴하는 함수를 만든다.

```python
def rouge_score(answer: str, prediction: str, rouge_types: list = ['rouge1', 'rouge2', 'rougeL']) -> dict:
    # ROUGE 지표의 의미
    # rouge1: 단어 하나하나가 얼마나 겹치나 본다. 정보의 포함 여부
    # rouge2: 두 단어씩 묶어서 얼마나 겹치나 본다. 문구의 흐름 확인
    # rougeL: 가장 길게 일치하는 문장 성분을 본다. 전체적인 구조 확인
    # ROUGE 지표 유효성을 검사한다. 사용자가 rouge_types으로 넘겨준 리스트에서 실제 계산 가능한 유형만 골라낸다.
    valid_rouge_type = ['rouge1', 'rouge2', 'rougeL']
    rouge = [rt for rt in rouge_types if rt in valid_rouge_type]
    # print(rouge)
    # 만약에 ROUGE 지표 유효성을 검사 후 유효한 지표가 하나도 없다면, 인위적으로 에러를 발생시켜 실행을 중단한다.
    if not rouge:
        raise ValueError('유효한 ROUGE 유형이 제공되지 않았습니다.')
        
    # RougeScorer의 인수로 계산할 ROUGE type과 한국어의 경우 조사나 어미가 달라도 핵심 의미가 같으면 같은 단어로 인식하는 여부를 지정해서 RougeScorer를
    # 계산하는 객체를 생성한다.
    rouge_score = rouge_scorer.RougeScorer(
        # 계산할 ROUGE type을 지정한다.
        rouge_types=rouge,
        # use_stemmer 속성의 속성값을 True로 지정하면 조사나 어미가 달라도 핵심 의미가 같으면 같은 단어로 인식한다.
        use_stemmer=True
    )
    # score() 메소드의 인수로 실제 정답과 LLM이 생성한 답변을 넘겨서 RougeScorer를 계산한다.
    # score() 메소드를 실행하면 내부적으로 precision(정밀도), recall(재현율), fmeasure(f1-score)가 모두 산출된다.
    score = rouge_score.score(answer, prediction)
    # print(score)
    
    # 보통 ROUGE 지표를 대표값으로 쓸 때 가장 많이 사용하는 계산된 값들 중 정밀도와 재현율의 조화 평균인 fmeasure만 리턴시킨다.
    return {rouge_type: score[rouge_type].fmeasure for rouge_type in rouge}
```

```python
print('질문:', question) # 사용자가 던진 원래 질문
print('답변:', answer) # 실제 정답
print('OpenAI 모델 예측:', openai_prediction) # LLM이 생성한 답변

score = rouge_score(answer, openai_prediction, ['rouge1'])
# print(score)
print(f'ROUGE1 지표 점수: {score["rouge1"]:.5f}')
```
