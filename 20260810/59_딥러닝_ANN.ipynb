인공 신경 망(Artifical Neural Network, ANN)

인공 신경망의 개념

다층 퍼셉트론(multi layer perceptron)은 이름 그대로 퍼셉트론의 층이 여러개라는 뜻으로 다층 퍼셉트론은 기존의 데이터 공간을 변형함으로써 기존 하나의 퍼셉트론으로 해결할 수 없었던 문제를 해결할 수 있게 되는 것이다. 이처럼 다수의 뉴런을 사용해서 만든 것은 인공 신경망이라고 한다.

<img src="./ann01.png" width="500" align="left" />

입력층(input layer)에 입력된 데이터는 출력층(output layer)에 도달하기 전에 은닉층(hidden layer)이라는 층을 거친 후 출력층에 도달한다. 은닉층은 하나의 층뿐만 아니라 다수의 층으로 정할 수 있고 딥러닝이라는 이름은 이 은닉층의 깊이가 깊다는 뜻(함수가 깊게 중첩된다.)에서 나온 이름이다. 입력층의 노드 개수는 피쳐 개수와 동일하고 출력층의 노드 개수는 분류하려는 클래스 개수와 같다.

출력층의 각 노드나 나타내는 숫자는 해당 클래스에 대한 점수(확률)이며 점수가 높을수록 해당 클래스에 속할 확률이 높다는 의미이고 최종적으로 점수가 가장 높은 클래스를 선정한다.

신경망 함수 $f(x) = f_3(f_2(f_1(x)))$와 같은 합성 함수 형태로 표현된다. $f_1$을 첫 번째 신경망 층, $f_1$를 두 번째 신경망 층, $f_3$을 세 번째 신경망 층이라고 생각하면 된다. 신경망 층이 깊을수록 함수의 개수는 많아지고 합성 함수도 복잡해진다. 이때 합성 함수의 개수는 모델의 깊이(depth)를 의미하는데, 딥러닝에서 deep이라는 용어가 바로 여기서 나온 용어이다.

오차 역전파(back propagation)

다층 퍼셉트론에서 최적의 값을 찾아가는 과정은 오차 역전파 방법을 사용한다.

<img src="./ann02.png" width="500" align="left" />

입력층 → 은닉층 → 출력층 순서대로 흘러가는 것을 순전파(forward propagation)라 하고 반대로 역전파는 출력층 → 은닉층 → 입력층 순서대로 반대로 거슬러 올라가는 방향이다. 오차 역전파를 기반으로 가중치와 편향값을 수정한 후 더 좋은 성능을 낼 수 있도록 모델을 개선한다.

활성화 함수(activation function)

활성화 함수는 딥러닝에서 입력값과 가중치 및 편향값을 계산해서 해당 노드를 활성화 할지를 결정하는 함수이다. 딥러닝에서는 결과값을 결정하는 여러가지의 활성화 함수가 존재한다.

① `항등 함수(identity function)`  
항등 함수 또는 선형 함수(linear function)라고도 부르는 이 함수는 입력과 출력이 같은 값을 가진다. 주로 회귀 문제에서의 활성화 함수로 사용된다.

항등 함수는 아래의 식을 따른다.

$$\phi(x) = x$$

항등 함수를 그림으로 나타내면 아래와 같다.

② `계단 함수(step function)`  
계단 함수는 입력값이 0 이하일 경우 0을 출력하고, 0을 초과할 때만 1을 출력한다. 계단 함수의 출력은 0 또는 1로 오직 두 가지 값만 가진다는 특징이 있다.

계단 함수는 아래의 식을 따른다.

$$\phi(x) = 
\begin{cases}
0, \; x \leq 0 \\
1, \; x > 0
\end{cases}$$

계단 함수를 그림으로 나타내면 아래와 같다.

<img src="./ann03.png" width="400" align="left" />

입력값 $x$가 0을 기준으로 넘기 전과 후의 값에 따라 출려값이 바뀌는 것을 알 수 있다. 계단 함수는 사용하기 간단하다는 장점이 있지만 미분이 불가능하다는 단점이 있다.

③ `부호 함수(sign function)`  

부호 함수는 아래의 식을 따른다.

$$\phi(x) = 
\begin{cases}
-1, \; x < 0 \\
0, \; x = 0 \\
1, \; x > 0
\end{cases}$$

부호 함수를 그림으로 나타내면 아래와 같다.

<img src="./ann04.png" width="400" align="left" />

계단 함수와 비슷하게 생겼지만 0 또는 1만 출력하는 계단 함수와는 달리 -1, 0, 1 값을 출력하는 것을 알 수 있다.

④ `시그모이드 함수(sigmoid function)`

시그모이드 함수는 아래의 식을 따른다.

$$\phi(x) = \frac {1} {1 + e^{-x}}$$

시그모이드 함수를 그림으로 나타내면 아래와 같다.

<img src="./ann05.png" width="400" align="left" />

시그모이드 함수는 0과 1사이의 값을 출력한다.  
시그모이드 함수를 딥러닝에 적용했을 때 단점은 학습하는 과정에서 미분을 반복하면 기울기 값이 매우 작아지는 기울기 소실(vanishing gradiend) 문제가 발생할 수 있다는 것이다. 시그모이드 함수에서 $x$ 값이 지나치게 크거나 작을 경우, 미분값이 0에 가까워지고 이는 기울기 소실 문제를 발생시켜 학습 속도가 늘려질 수 있다.

⑤ `렐루 함수(Rectified Linear Function, ReLU)`

렐루 함수는 아래의 식을 따른다.

$$\phi(x) = 
\begin{cases}
0, \; x \leq 0 \\
x, \; x > 0
\end{cases}$$

위의 식은 아래와 같이 한 줄로도 표현할 수 있다.

$$\phi(x) = max(x, 0)$$

렐루 함수를 그림으로 나타내면 아래와 같다.

<img src="./ann07.png" width="500" align="left" />

렐루 함수는 $x$값이 0 이하라면 0을 출력하고, $x$값이 양수면 $x$값을 그대로 출력한다. 계단 함수, 부호 함수, 시그모이드 함수 함수와는 다르게 출력값의 상한선이 없다는 특징이 있다.

⑥ `리키 렐루 함수(Leaky ReLU Function)`

리키 렐루 함수는 아래의 식을 따른다.

$$\phi(x) = 
\begin{cases}
ax, \; x \leq 0 \\
x, \; x > 0
\end{cases}$$

위의 식은 $a \leq 1$ 이라면 아래와 같이 한 줄로도 표현할 수 있다.

$$\phi(x) = max(x, ax)$$

리키 렐루 함수를 그림으로 나타내면 아래와 같다.

<img src="./ann08.png" width="500" align="left" />

⑦ `하이퍼볼릭 탄젠트 함수(Hyperbolic Tangent Function, tanh)`

하이퍼볼릭 탄젠트 함수는 아래의 식을 따른다.

$$\phi(x) = \frac {e^x - e^{-x}} {e^x + e^{-x}}$$

시그모이드 함수와의 관계를 통해 아래와 같이 표현할 수도 잇다.

$$\phi(x) = 2 \times sigmoid(2x) - 1$$

하이퍼볼릭 탄젠트  함수를 그림으로 나타내면 아래와 같다.

<img src="./ann06.png" width="500" align="left" />

시그모이드 함수와 비슷하지만 시그모이드 함수의 출력 범위는 0부터 1사이인 반면에 하이퍼볼릭 탄젠트 함수의 출력 범위는 -1부터 1사이이다.

⑧ `소프트맥스 함수(softmax function)`  

소프트맥스 함수는 주로 최종 출력층에 사용되는 활성화 함수이다. 만약헤 해결해야 할 문제가 회귀 문제라면 출력층에 항등 함수를 사용하고 분류 문제일 경우 소프트맥스 함수를 사용한다.

소프트맥스 함수는 아래와 같은 식을 따른다.

$$\phi(x) = \frac {exp(x_k)} {\sum_{i=1}^n exp(x_i)}$$

소프트맥스 함수를 사용할 때 위의 식을 그대로 사용할 경우 오버플로우 문제가 발생될 수 있다. 오버플로우는 출력값이 컴퓨터가 표현할 수 있는 수의 한계를 초과하는 문제를 말한다. 예를 들어 $x_k$ = 1000일 때, exp(1000)는 매우 큰 수가 되므로 계산하는 것이 불가능하다. 이 문제를 해결하기 위해 소프트맥스 함수를 아래와 같이 변형해서 쓰기도 한다.

$$\phi(x) = \frac {exp(x_k + C)} {\sum_{i=1}^n exp(x_i + C)}$$

소프트맥스 함수에서 지수 연산을 할 때, 입력값에 임의의 상수 $C$를 더하거나 빼서 오버플로우 문제를 해결한다. 임의의 상수 $C$는 일반적으로 입력값의 최대값을 이용한다. 만약에 입력값이 999, 1000, 1001이라 하면 입력값의 최대값은 1001이므로 임의의 상수 $C$는 1001이 된다. exp(999), exp(1000), exp(1001)를 구해야 해서 오버플로우가 발생할 수 있지만, 입력값의 최대값을 빼주면 exp(-2), exp(-1), exp(0)와 같이 연산이 편해진다.

배치 정규화(batch normalization)

배치 정규화는 해당층의 값의 분포를 변경하는 방법으로, 평균과 분산을 고정시키는 방법이다. 배치 정규화 이용하면 기울기 소실을 줄임으로써 신경망 학습 속도를 향상시킬 수 있는 장점이 있다.

드롭아웃(dropout)

신경망의 모든 노드를 사용하는 것이 아닌 일부 노드를 사용하는 방법이다.

<img src="./ann11.png" width="1000" align="left" />

왼쪽 그림은 드롭아웃을 적용하기 전 신경망이고, 오른쪽 그림은 드롭아웃을 적용한 신경망이다.

드롭아웃 신경망에서 노드를 일시적으로 제거하는 방법이다. 어떤 노드를 신경망에서 제거할지는 각 계층에서 무작위로 선택된다. 드롭아웃을 적용하면 신경망의 노드 숫자가 줄어들고 이에 따라 연산량도 줄어들고 과적합을 방지할 수 있다는 장점이 있다.

tensorflow 2.x

tensorflow는 파이썬을 이용해서 딥러닝 학습시 사용하는 라이브러리이다. tensorflow를 활용하면 신경망을 기본으로 하는 딥러닝에서 다양한 신경망 관련 연산을 처리할 수 있어 편리하다.

tensorflow를 이용해서 신경망 구조를 만드는 방법은 크게 `시퀀스`를 사용하는 방법과 `함수`를 사용하는 방법이 있다.

`시퀀스`를 사용하는 방법

Sequential 클래스 객체를 선언하고 Sequential 모델에 add() 메소드로 layer를 추가해서 쌓아올린다.

```python
# layer들을 일렬로 차례대로 쌓아올리는 순차적 모델 구조를 생성하기 위해 import 한다.
from tensorflow.keras.models import Sequential
# 신경망의 가장 기본이 되는 Dense Layer(Fully Connected, 완전 연결층, 밀집층)를 생성하기 위해 import 한다.
from tensorflow.keras.layers import Dense
```

```python
# 신경망 layer를 순차적으로 쌓아올리기 위한 비어있는 Sequential 모델을 만든다.
model = Sequential()

# 첫 번째(입력) layer인 Dense layer를 만들어 Sequential 모델에 추가한다.
# input_shape 속성은 입력 데이터 1건의 형태를 지정하며 첫 번째 층에서만 필수로 요구된다. 가로 32, 세로 32, 채널 1인 흑백 이미지가 입력됨을 의미한다.
# units 속성은 이 층이 다음 층으로 내보낼 출력 데이터의 차원을 지정한다. 뉴런의 개수
# activation 속성은 layer에서 사용할 활성화 함수를 지정한다.
model.add(Dense(input_shape=(32, 32, 1), units=100, activation='relu'))

# 히든 layer인 Dense layer를 만들어 Sequential 모델에 추가한다.
# 첫 번째 layer는 입력 데이터의 차원을 지정하는 input_shape 속성이 필요하지만 첫 번째 layer를 제외한 나머지 layer는 이전 layer의 출력이 현재 layer의 입력으로
# 들어오기 때문에 input_shape 속성이 필요없다.
model.add(Dense(units=50, activation='relu'))

# 마지막(출력) layer
# 소프트맥스 함수는 5개 뉴런의 출력값을 모두 합치면 1.0(100%)이 되도록 확률값으로 변환해준다.
model.add(Dense(units=5, activation='softmax'))

# summary() 메소드로 만들어진 모델의 구조를 확인할 수 있다.
model.summary()
# Dense layer의 Param은 '(입력 개수 + 바이어스) * 출력 개수'로 계산되고 바이어스는 무조건 layer당 1개 이다.
```

`함수`를 사용하는 방법

```python
# 입력과 출력을 지정하여 전체 모델 객체를 생성하기 위해서 import 한다.
from tensorflow.keras.models import Model
# Input: 모델에 들어오는 입력 데이터 형태를 정의하기 위해서 import 한다.
# Activation: layer의 연산 결과에 활성화 함수를 적용하기 위해서 import 한다.
from tensorflow.keras.layers import Input, Dense, Activation

# 모델에 입력될 데이터의 형태를 정의한다.
input_layer = Input(shape=(32, 32, 1))

# 첫 번째(입력) layer
# layer = Dense(units=100)(input_layer) # input_layer 입력으로 받는 100개 뉴런의 첫 번째 Dense layer를 만든다.
# layer = Activation('relu')(layer) # 첫 번째 Dense layer의 출력에 relu 활성화 함수를 적용한다.
layer = Dense(units=100, activation='relu')(input_layer)

# 히든 layer
# layer = Dense(units=50)(layer) # layer를 입력으로 받는 50개 뉴런의 히든 Dense layer를 만든다.
# layer = Activation('relu')(layer) # 히든 Dense layer의 출력에 relu 활성화 함수를 적용한다.
layer = Dense(units=50, activation='relu')(layer)

# 마지막(출력) layer
# output_layer = Dense(units=5)(layer) # layer를 입력으로 받는 5개 뉴런의 마지막 Dense layer를 만든다.
# output_layer =  Activation('softmax')(output_layer) # 출력 Dense layer의 출력에 softmax 활성화 함수를 적용한다.
output_layer = Dense(units=5, activation='softmax')(layer)

# Model 객체에 입력 데이터의 형태와 출력 레이어를 넘겨서 모델을 만든다.
model = Model(input_layer, output_layer)

model.summary()
```
