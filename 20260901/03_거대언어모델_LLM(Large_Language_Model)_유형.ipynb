# LangChain 모델 유형

랭체인 문서에 따르면 LLM과 Chat Model 클래스는 각각 다른 형태의 입력과 출력을 다루는 언어 모델을 나타낸다. 이 두 모델은 각기 다른 특성과 용도를 가지고 있어, 사용자의 요구사항에 맞게 선택하여 사용할 수 있다. 일반적으로 LLM은 주로 단일 요청에 대한 복잡한 출력을 생성하는 데 적합한 반면, Chat Model은 사용사와의 상호작용을 통한 연속적인 대화 관리에 더 적합하다.

## LLM

LLM 클래스는 텍스트 문자열을 입력으로 받아 처리한 후, 텍스트 문자열을 반환한다. 이 모델은 광범위한 언어 이해 및 텍스트 생성 작업에 사용된다.  
예를들어, 문서 요약, 콘테츠 생성, 질문에 대한 답변 생성 등 복잡한 자연어 처리 작업을 수행할 수 있다. 사용자가 특정 주제에 대한 설명을 요청할 때, LLM은 주어진 텍스트 입력을 바탕으로 상세한 설명을 생성하여 반환한다.

## Chat Model

Chat Model 클래스는 메시지의 리스트를 입력으로 받고, 하나의 메시지를 반환한다. 이 모델은 대화형 상황에 최적화되어 있으며, 사용자와의 연속적인 대화를 처리하는 데 사용된다.  
Chat Model은 대화의 맥락을 유지하면서 적절한 응답을 생성하는 데 중점을 두고 사용자가 챗봇과 대화하는 상황에서, 사용자의 질문과 이전 대화 내용을 고려하여 적절한 답변을 생성한다.

# OpenAI

LangChain에서 OpenAI와 ChatOpenAI는 모두 OpenAI의 모델을 호출하기 위한 클래스이지만, 호출하는 모델의 종류와 입출력 구조에서 차이가 있습니다.

`구분              OpenAI                                             ChatOpenAI                              `  
`대상 모델         텍스트 완성형 모델(Legacy Completion Models)       대화형 모델(Chat Models)                `  
`                   (예: gpt-3.5-turbo-instruct, text-davinci-003)     (예: gpt-4o, gpt-4o-mini, gpt-4-turbo)   `  
`입력 형태         단일 문자열(String)                                메시지 객체 리스트(System, Human, AI 등)`  
`출력 형태         단일 문자열(String)                                메시지 객체(AIMessage)                  `  
`권장 프롬프트     PromptTemplate                                     ChatPromptTemplate                      `  
`현재 위상         구버전 모델 호환용(Legacy)                         표준 및 기본 사용 클래스                `

단순히 문자열을 입력받아 그 뒤에 이어질 텍스트를 완성하여 문자열로 반환하기 위해 import 한다.

System, Human 같은 역할 개념이 없으며, 주어진 프롬프트를 이어 쓰는 형태로 동작합니다.

```python
from langchain_openai import OpenAI
```

```python
llm = OpenAI()
print(llm)
print('-' * 100)

result = llm.invoke('한국의 대표적 관광지 3군데를 추천해줘.')
result
```

텍스트 완성형 모델인 OpenAI는 정책이 변화되면서 모델 지원이 종료될 것으로 예상된다.

# ChatOpenAI

이전의 OpenAI 클래스와 달리 대화형 AI에 최적화된 방식이다.

```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI()
print(llm)
print('-' * 100)

result = llm.invoke('한국의 대표적 관광지 3군데를 추천해줘.')
result
```

LCEL을 사용하여 AI에게 '여행 전문가'라는 정체성을 부여하고 한국의 대표적 관광지 3군데를 추천받는 모델을 만든다.

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_messages([
    ('system', '이 시스템은 여행 전문가입니다.'),
    ('user', '{user_input}')
])

llm = ChatOpenAI(model='gpt-4o-mini')

chain = prompt | llm

result = chain.invoke({'user_input': '안녕하세요? 한국의 대표적인 관광지 3군데를 제주도 빼고 추천해주세요.'})
```
