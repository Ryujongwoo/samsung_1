# LangChain의 모델 파라미터 설정

모델의 기본 속성값을 조정하는 방법에 대해서 살펴본다.  
모델의 속성에 해당하는 모델 파라미터는 모델의 출력을 조정하고 최적화하는데 사용되며, 모델이 생성하는 텍스트의 스타일, 길이, 정확도 등에 영향을 준다.

```python
import os

# os.environ['OPENAI_API_KEY'] = '사용자 API Key'
```

```python
from langchain_openai import ChatOpenAI
```

## ChatOpenAI에 적용하는 주요 파라미터

`model`: 사용할 gpt 모델을 지정한다.  
`temperature`: 모델이 응답을 생성할 때 얼마나 창의적이거나 결정론적일지 제어한다. 0.0에서 2.0 사이의 실수값을 가진다. 보통 0.7 전후로 설정한다.  
`max_completion_tokens`: 생성할 최대 토큰 수를 지정해서 생성할 텍스트의 길이를 제한한다.  
`frequency_penalty`: 값이 클수록 이미 등장한 단어나 구절이 다시 등장할 확률을 감소시킨다. 반복을 줄이고 텍스트의 다양성을 증가시킨다. 0 ~ 1 사이의 실수값  
`presence_penalty`: 단어의 존재 유무에 따라 그 단어의 선택 확률을 저정한다. 값이 클수록 아직 등장하지 않은 새로운 단어가 사용된다. 0 ~ 1 사이의 실수값  
`stop_sequences`: 특정 단어나 구절이 등장할 경우 생성을 멈추도록 설정해서 출력을 특정 포인트에서 종료하고자 할 때 설정한다.

```python
# 모델 객체 생성 및 파라미터 설정
llm = ChatOpenAI(
    model='gpt-4o-mini', # 모델 종류: 성능과 비용의 밸런스가 좋은 소형 모델을 지정한다.
    temperature=0.7, # 창의성: 0에 가까워질 수록 일관된 답변을 하고, 2에 가까워질 수록 창의적(무작위)인 답변을 한다. 0.7은 적당히 유연한 답변을 생성한다.
    max_completion_tokens=100, # 길이 제한: 생성될 답변의 최대 길이를 토큰 단위로 제한한다.
    frequency_penalty=0.5, # 빈도 페널티: 이미 사용한 단어를 다시 사용할 때 감점을 주어 '똑같은 말 반복'을 방지한다.
    presence_penalty=0.5, # 존재 패널티: 새로운 소재나 단어를 언급할 때 가산점을 주어 '다양한 주제'를 다루도록 유도한다.
    stop_sequences=['\n'] # 정지 시퀀스: AI가 답변 도중 '줄바꿈(\n)'을 하려고 하면 즉시 생성을 멈춘다.
)

# 모델 실행
# 모델 객체를 생성할 때 설정한 temperature나 stop_sequences 등의 옵션이 적용되어 답변 생성된다.
# result = llm.invoke('태양계에서 가장 큰 행성은 무엇인가요?')

# 질문 정의
# AI에게 보낼 단순 문자열 형태의 프롬프트를 변수에 저장한다.
question = '태양계에서 가장 큰 행성은 무엇인가요?'
result = llm.invoke(input=question)
```

```python
result
```

## 모델 실행시 파라미터 전달하기

이미 생성된 모델을 invoke() 메소드로 실행할 때 특정 파라미터를 덮어씌워 실행할 수 있다.

```python
# invoke() 메소드가 실행되는 시점에서 ChatOpenAI에 모델을 만들 때 지정한 파라미터를 무시하고 다시 지정할 수 있다.
# 처음에 ChatOpenAI에 모델을 만들 때 설정했던 값보다 invoke() 메소드에서 지정하는 파라미터가 우선순위를 가진다.
result = llm.invoke(input=question, max_completion_tokens=10)
```

```python
result
```

모델 호출 시 적용할 파라미터들을 딕셔너리 형태로 임시 파라미터로 설정한다.

```python
# 임피 파라미터 설정
# 임시 파라미터로 설정한 내용을 invoke() 메소드가 실행될때 앞에 '**'을 앞에 붙여서 사용하면 딕셔너리의 내용을 펼쳐서(unpacking) 메소드에 전달한다.
params = {
    'max_completion_tokens': 20
}

result = llm.invoke(input=question, **params)
```

```python
result
```

## 모델에 추가 파라미터 전달하기

bind() 메소드를 사용해서 체인에 새로운 파라미터를 추가해서 연결해서 다양한 상황에 맞게 모델의 동작을 제어할 수 있다.

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate([
    ('system', '이 시스템은 천문학 질문에 답변할 수 있습니다.'),
    ('user', '{user_input}')
])

llm = ChatOpenAI(model='gpt-4o-mini', max_completion_tokens=100)
# chain이 없을 때 모델에서 invoke() 메소드 실행하기
message = prompt.format_messages(user_input='태양계에서 가장 큰 행성은 무엇인가요?')
before_answer = llm.invoke(message)
before_answer.content
```

```python
chain = prompt | llm
# chain이 있을 때 모델에서 invoke() 메소드 실행하기
before_answer = chain.invoke({'user_input': '태양계에서 가장 큰 행성은 무엇인가요?'})
before_answer.content
```

기존 모델 객체는 그대로 둔 채, 이 체인 내에서만 사용될 모델의 파라미터 설정을 bind() 메소드로 묶어준다.

```python
chain = prompt | llm.bind(max_completion_tokens=10)
after_answer = chain.invoke({'user_input': '태양계에서 가장 큰 행성은 무엇인가요?'})
after_answer.content
```
