```python
import warnings
warnings.filterwarnings("ignore")
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
import numpy as np
```

장단기 기억 모델(Long Short Term Memory, LSTM)

LSTM은 순차 데이터를 처리하기 위해 고안된 순환 신경망(Recurrent Neural Network, RNN)의 일종으로 기존 RNN이 가진 치명적인 한계를 극복하기 위해 개발되었으며, 현재도 자연어 처리(NLP), 음성 인식 등 다양한 분야에 사용된다.

LSTM은 RNN에서 발생되는 단점인 기울기 소실(gradient vanishing) 또는 기울기 폭주(gradient exploding)와 같은 RNN의 단점을 극복하고자 만들어진 좀 더 진화된 RNN으로 LSTM 셀 내부를 살펴보면 기존 RNN의 단점을 극복하기 위해 이전 정보를 지우거나 기억하기 위한 로직과 현재 정보를 기억하기 위한 로직이 추가로 구현되어 있다.

아래 그림을 살펴보면 RNN 셀에서 보지못했던 주황색 선과 조금 더 많아진 활성화 함수와 수학 기호들을 볼 수 있다.

주황색 선은 메모리 셀이라 부른다.  
주황색 선의 곱하기 기호에서 시그모이드의 출력값이 곱해지게 되어 메모리 셀의 기존 정보를 어디까지 기억할지 결정하게 된다.  
주황색 선의 더하기 기호에서 새로운 정보와 시그모이드의 출력값이 곱해지에 되어 새로운 정보를 어느 정도까지 메모리 셀의 기존 정보에 더할지 결정하게 된다.  
h<sub>t</sub> 선상의 곱하기 기호에서 메모리 셀의 정보와 새로운 장보가 계산되서 출력(상태)값을 출력한다.

tensorflow에서 이미 LSTM이 구현되어 있기 때문에 위 그림을 직접 구현할 필요는 없고 아래 그림만 이해해도 사용하는데 큰 무리는 없다.

```python
inputs = np.array([[[1, 2]]])
lstm_inputs = tf.constant(inputs, dtype=tf.float32)
sess = tf.Session()
print('입력 데이터: {}'.format(sess.run(lstm_inputs)))
```

LSTM 셀을 만든다.

```python
# LSTMCell() 객체를 생성할 때 num_units 속성값으로 RNN 셀의 개수를 지정해서 생성한다.
lstm_cell = tf.nn.rnn_cell.LSTMCell(num_units=1)
print('LSTM 셀의 개수: {}'.format(lstm_cell.state_size))
outputs, state = tf.nn.dynamic_rnn(cell=lstm_cell, inputs=lstm_inputs, dtype=tf.float32)
print('출력값: {}'.format(outputs))
print('상태값: {}'.format(state))
```

LSTM의 출력값, 상태값, 메모리 셀값을 확인한다.

```python
with tf.Session() as sess:
    sess.run(tf.global_variables_initializer())
    
    _outputs, _state = sess.run([outputs, state])
    print('출력값: {}'.format(_outputs))
    print('상태값: {}'.format(_state.h))
    print('메모리 셀: {}'.format(_state.c))
```
