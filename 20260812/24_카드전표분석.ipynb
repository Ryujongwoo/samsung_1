# 효율적인 프로젝트 진행을 위한 프로젝트 로드맵

## 프로젝트 로드맵은 크게 세 단계로 나눌 수 있다.

① 프로젝트 목표를 수립한다.  
&nbsp;&nbsp;&nbsp;▶ 프로젝프 목표는 프로젝트를 통해 얻고자 하는 결과가 무엇인지 명확하게 드러나야 한다.  
&nbsp;&nbsp;&nbsp;▶ 목표를 수립하고 나면 프로젝트를 진행할 데이터를 선정해야 한다.  
&nbsp;&nbsp;&nbsp;▶ 프로젝트 목표 수립 단계에서 목표 수립과 데이터 선정 순서가 바뀔 수도 있다. 데이터를 먼저 수집하고, 수집된 데이터를 바탕으로 목표를 수립할 수도 있다.

② 프로젝트 구현한다.  
&nbsp;&nbsp;&nbsp;▶ 프로젝트 목표를 실제 동작하는 프로그램으로 만드는 단계  
&nbsp;&nbsp;&nbsp;▶ 코드를 작성하기 전에 알고리즘을 작성하는데 알고리즘 어떤 문제를 해결하기 위해 필요한 절차를 순서대로 표현한 것 이다.  
&nbsp;&nbsp;&nbsp;▶ 알고리름은 문제, 해결, 절차라는 3가지 키워드만 기억하면 되고 이 3가지 키워드로 알고리즘을 설명할 수 있다.

③ 프로젝트 결과를 분석한다.  
&nbsp;&nbsp;&nbsp;▶ 마지막은 결과 분선 단계이다. 프로그램을 실행해 어떤 결과가 나오면 이 결과를 분석하고 정리해서 목표를 달성하는 단계이다.

csv 파일에서 데이터를 읽어들인다.

데이터 분석에서 사용하는 데이터 파일은 여러 종류가 있다. xls, xlsx 형식의 엑셀 파일도 있고, 텍스트로만 이루어진 파일이 있다. 확장명이 txt인 파일이 아니고 csv나 tsv인 파일을 말한다. csv 파일은 쉼표로 데이터가 구분된 파일이고 tsv 파일은 tab(`\t`)으로 데이터가 구분된 파일이다.

```python
import csv
import matplotlib.pyplot as plt
import koreanize_matplotlib
```

```python
# open() 함수를 사용해서 card.csv 파일을 읽기 전용으로 open 한다.
file = open('./data/card.csv', encoding='utf8')
# print(file)
# reader() 메소드로 open 함수로 open한 파일의 내용을 읽어온다.
data = csv.reader(file)
# print(data)
# next() 함수는 일어들인 csv 파일에서 한 행을 건너뛴다.
next(data)
# 읽어들인 csv 파일의 내용을 list() 생성자 함수를 실행해서 리스트로 변환한다.
data = list(data)
# print(type(data))
print(data)
```

필요한 데이터만 추출하기

```python
# 2019-10-12에 네이버파이낸셜(주)에서 546,800원 결제
for row in data:
    date = row[0].split()[0]
    store = row[5]
    price = int(row[6])
    print('{} {}에서 {:,d}원 결제'.format(date, store, price))
```

구매 확정 확인하기

```python
for row in data:
    if row[-1] == '전표매입':
        print('구매확정')
    else:
        print('구매취소')
```

카드 이용내역 분석하기

총 지출액 계산하기

```python
total = 0
for row in data:
    if row[-1] == '전표매입':
        price = int(row[6])
        total += price
print(total)
```

월별 지출액 시각화

```python
total = [0, 0, 0]
for row in data:
    if row[-1] == '전표매입':
        month = int(row[0].split()[0].split('-')[1])
        price = int(row[6])
        total[month - 10] += price
print(sum(total))
print(total)
```

```python
import matplotlib.ticker as ticker

plt.figure(figsize=(8, 6))
plt.title('10 ~ 12월 지출현황')
plt.bar(['10월', '11월', '12월'], total, color='royalblue')
plt.ticklabel_format(style='plain', axis='y') # 지수 형태 사용하지 않기
plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}')) # 천 단위마다 ',' 출력하기
plt.show()
```

지출액 비교 시각화

```python
taxi = [0, 0, 0]
for row in data:
    # if row[-1] == '전표매입' and row[5].find('택시') >= 0:
    if row[-1] == '전표매입' and '택시' in row[5]:
        month = int(row[0].split()[0].split('-')[1])
        price = int(row[6])
        taxi[month - 10] += price
print(sum(taxi))
print(taxi)
```

```python
plt.figure(figsize=(8, 6))
plt.title('10 ~ 12월 택시비 현황')
plt.plot(['10월', '11월', '12월'], taxi, color='crimson')
plt.ticklabel_format(style='plain', axis='y')
plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))
plt.show()
```

```python
food = [0, 0, 0]
for row in data:
    if row[-1] == '전표매입' and row[5] == '(주)우아한형제들':
        month = int(row[0].split()[0].split('-')[1])
        price = int(row[6])
        food[month - 10] += price
print(sum(food))
print(food)
```

```python
plt.figure(figsize=(8, 6))
plt.title('10 ~ 12월 배달음식비 현황')
plt.plot(['10월', '11월', '12월'], food, color='indigo')
plt.ticklabel_format(style='plain', axis='y')
plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.title('10 ~ 12월 택시비/배달음식비 현황')
plt.plot(['10월', '11월', '12월'], taxi, color='crimson', label='택시비')
plt.plot(['10월', '11월', '12월'], food, color='indigo', label='배달음식비')
plt.ticklabel_format(style='plain', axis='y')
plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))
plt.legend()
plt.show()
```

지출액 상위 10개 시각화

```python
import operator
```

```python
expend = {}
for row in data:
    if row[-1] == '전표매입':
        store, price = row[5], int(row[6])
        # 가맹점명이 가맹점명과 이용금액을 기억하는 expend라는 딕셔너리의 key 목록에 포함되어 있지 않은가?
        if store not in expend.keys():
            # 포함되어 있지 않다면 가맹점명을 key로하고 이용금액을 value로 하는 데이터를 expend 딕셔너리에 추가한다.
            expend[store] = price
        else:
            # 포함되어 있다면 가맹점명에 이용금액을 더해준다.
            expend[store] += price
# print(expend)

top10 = sorted(expend.items(), key=operator.itemgetter(1), reverse=True)[:10]
top10
```

```python
top10_store = []
top10_price = []

for key, value in top10:
    top10_store.append(key)
    top10_price.append(value)
print(top10_store, top10_price)
```

```python
plt.figure(figsize=(8, 6))
plt.title('10 ~ 12월 지출 top10 현황')
plt.plot(top10_store, top10_price, color='crimson')
plt.ticklabel_format(style='plain', axis='y')
plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.title('10 ~ 12월 지출 top10 현황')
plt.barh(top10_store, top10_price, color='crimson')
plt.ticklabel_format(style='plain', axis='x')
plt.gca().xaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))
plt.xticks(rotation=45)
plt.show()
```
