# LangChain에서 사용할 도구 생성하기 - Tavily 웹 검색 도구

LangChain은 검색, 코드 인터프리터, 생산성 도구 등 다양한 Tool(도구)를 직접적 또는 제휴 형태로 제공한다.

Tavily를 사용해서 웹 검색을 하기위해서 TavilySearchResults를 import 한다.

```python
from langchain_community.tools import TavilySearchResults
```

```python
query = '스테이크와 어울리는 와인을 추천해주세요.'
```

Tavily 웹 검색 도구를 정의한다.

```python
# TavilySearchResults 클래스의 생성자로 웹을 검색해서 가져올 문서의 개수를 넘겨서 Tavily 웹 검색 도구를 정의한다.
web_search = TavilySearchResults(max_results=2)
```

위의 경고 메시지는 TavilySearchResults 클래스가 이전 패키지(langchain-community)에서 새로 분리된 전용 패키지(langchain-tavily)로 이관되었기 때문에 발생한다.

이관된 방식으로 Tavily를 사용하고 싶다면 `pip install langchain-tavily`를 실행해서 라이브러리를 설치하고 import 하고 사용한다.  

`from langchain_tavily import TavilySearch`  
`web_search = TavilySearch(max_results=2)`

Tavily 웹 검색 도구로 웹 검색을 실행한다.

```python
# invoke()의 인수로 지정된 쿼리를 Tavily 서버에 검색 명령을 보내서 웹 검색을 실행한다.
search_results = web_search.invoke(query)
```

```python
print(type(search_results))
print(type(search_results[0]))
print(search_results[0].keys())
```

```python
for result in search_results:
    print(result)
    print('-' * 100)
```

```python
for result in search_results:
    print(result['content'])
    print('-' * 100)
```

```python
# 간단 텍스트 반환
duckduckgo_search = DuckDuckGoSearchRun()
duckduckgo_results = duckduckgo_search.invoke(query)
print(type(duckduckgo_results))
print(duckduckgo_results)
```

```python
# 링크/제목/요약 구조화 반환
duckduckgo_search = DuckDuckGoSearchResults(num_results=2, output_format='json')
duckduckgo_results = duckduckgo_search.invoke(query)
print(type(duckduckgo_results))

parsed_results = json.loads(duckduckgo_results)
print(type(parsed_results))
print(type(parsed_results[0]))
print(parsed_results[0].keys())
```

```python
for result in parsed_results:
    print(result)
    print('-' * 100)
```

```python
for result in parsed_results:
    print(result['snippet'])
    print('-' * 100)
```

생성한 web_search(Tavily 웹 검색 도구) 객체가 내부적으로 어떤 속성과 구조를 가지고 있는지 파악한다.

```python
print('자료형: ')
print(type(web_search))
print('-' * 100)

print('도구 이름: ')
# name 속성은 도구의 고유 식별 이름을 얻어온다.
# 도구 이름은 LLM이 여러 개의 도구 중 어떤 것을 사용할지 결정할 때 참조하는 이름이다.
print(web_search.name)
print('-' * 100)

print('설명: ')
# description 속성은 도구가 무엇을 하는 도구인지에 대한 설명을 얻어온다.
# 설명은 LLM이 이 설명을 읽고 '아~~~, 실시간 웹 검색이 필요할 때는 이 도구를 쓰면 되겠구나!'라고 판단하는 LLM에게 주는 지침서 역할을 한다.
print(web_search.description)
print('-' * 100)

print('스키마: ')
# schema() 메소드는 도구를 실행할 때 어떤 형식의 입력값이 필요한지 정의한 구조를 출력한다.
pprint(web_search.args_schema.schema())
print('-' * 100)
```
