# 가상환경 활성화

가상환경은 프로젝트마다 `격리된 환경`을 생성함으로써 프로젝트별 패키지를 관리한다.

파이썬은 같은 버전끼리 패키지를 공유하므로 프로젝트 A, 프로젝트 B에서 3.11 버전을 사용하고 있다면 프로젝트 A에서 설치한 패키지를 프로젝트 B에서 따로 설치할 필요없이 사용할 수 있다. 프로젝트마다 패키지를 매번 설치하지 않아되 되니 파이썬을 처음 입문한 경우 매우 편리하게 느껴진다.

하지만, 프로젝트마다 다른 버전의 패키지를 사용해야 한다면 치명적인 문제가 발생될 수 있다. 프로젝트 A는 판다스 2.X 버전을 사용해야 하고 프로젝트 B는 판다스 3.X 버전을 사용해야 한다면 같은 같은 파이썬 버전에서는 판다스 2.X 버전과 판다스 3.X 버전을 동시에 사용할 수 없기 때문에 에러가 발생할 수 있다.

가상환경을 사용하면 서로 다른 버전의 파이썬을 사용하거나 서로 같은 버전을 사용하더라도 `격리된 환경을 사용`하기 때문에 가상환경 마다 서로 다른 패키지를 설치해서 `안정적으로 패키지 버전을 관리`할 수 있다.

`conda activate 가상환경이름` 명령을 실행한다.  
가상환경이 활성화되면 프롬프트 앞에 `(가상환경이름)`이 표시된다. 기본환경(base)과 가상환경(python_basic)의 파이썬 버전이 다른 것을 확인할 수 있다.

<img src="./미니콘다설치_가상환경설정_12.png" width="1000" />

활성화된 가상환경을 비활성화 하려면 `conda deactivate` 명령을 실행한다.  
가상환경을 제거하려면 `conda remove --name 가상환경이름 --all` 명령을 실행한다.

# jupyter nbclassic 설치하기

파이썬에서 사용할 외부 패키지를 미니콘다에서 설치하는 방법은 `pip install 패키지이름`과 이용하는 방법과 `conda install 패키지이름`을 이용하는 방법이 있는데 `pip`는 패키지간 `의존성을 고려하지 않고 설치`하고 `conda`는 `패키지간 의존성을 고려해서 적절하게 설치`된다.

가상환경에서 작업할 폴더로 이동하고 `pip install nbclassic` 명령을 실행하면 nbclassic이 설치된다. 커스텀 환경 설정 방법은 python과 동일하다.

<img src="./미니콘다설치_가상환경설정_13.png" width="1000" />

nbclassic 실행은 가상환경에서 소스코드를 작업할 폴더로 이동한 후 일전에 만들어둔 `j.bat`를 실행하면 된다.

<img src="./미니콘다설치_가상환경설정_14.png" width="1000" />

설치된 패키지 목록을 확인하려면 `pip list` 또는 `conda list` 명령을 사용하면 된다.

# jupyter nbclassic에 가상환경 kernel 추가 및 삭제하기

`jupyter kernelspec list` 명령을 실행하면 현재 사용중인 커널 목록을 확인할 수 있다.  
`python3    C:\miniconda3\envs\python_basic\share\jupyter\kernels\python3`가 기본 커널이다.

<img src="./미니콘다설치_가상환경설정_15.png" width="1000" />

커널은 `python -m ipykernel install --user --name 가상환경이름` 명령을 실행해서 만들고 가상환경당 커널을 1개씩 만든다.  
만약에 ipykernel 때문에 에러가 발생되면 `pip install ipykernel` 명령을 실행해서 설치한 후 커널을 만들면 된다.

커널이 생성되면 nbclassic의 Kernel 메뉴의 Change Kernel에서 커널 목록을 확인할 수 있다.

<img src="./미니콘다설치_가상환경설정_16.png" width="400" />

커널을 삭제하려면 `jupyter kernelspec uninstall 커널이름` 명령을 실행한다.

<img src="./미니콘다설치_가상환경설정_17.png" width="1000" />
