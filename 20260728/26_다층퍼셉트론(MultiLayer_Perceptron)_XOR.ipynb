xor 연산은 입력 데이터를 X라고 할 경우 `[[0, 0], [0, 1], [1, 0], [1, 1]]`과 같은 경우의 수가 있다.  
총 4개의 입력 데이터가 각각 2개의 피쳐를 가지고 있으므로 X는 `[4, 2]`의 형태로 정의할 수 있다.  
입력값에 따른 레이블 Y는 `[[0], [1], [1], [0]]`이므로 Y는 `[4, 1]`의 형태로 정의할 수 있다.

```python
X = tf.placeholder(dtype=tf.float32, shape=[4, 2]) # 피쳐
Y = tf.placeholder(dtype=tf.float32, shape=[4, 1]) # 레이블
```

1번째 히든 레이어에서 2개의 입력(X)과 1개의 바이어스(B1)을 받아서 2개의 시그모이드 함수 출력(Z)을 2번째 히든 레이어로 전달한다.

2개의 X(x1, x2)와 바이어스(bias)를 입력으로 받는 2개의 뉴런을 만든다.

```python
# 2개의 입력을 받는 2개의 뉴런을 사용하기 위해서 가중치를 [입력 데이터의 개수, 뉴런의 개수] 만큼의 가중치를 난수를 만든다.
W1 = tf.Variable(tf.random_normal([2, 2], dtype=tf.float32))
# 각 뉴런은 1개의 바이어스를 가진다.
# tensorflow의 zeros() 메소드는 모든 원소의 값이 0인 텐서(2차원 이상의 행렬)를 생성한다.
B1 = tf.Variable(tf.zeros([2]))
# 시그모이드 함수를 이용해서 1번째 히든 레이어의 출력을 리턴한다.
Z = tf.sigmoid(tf.matmul(X, W1) + B1)
```

```python
sess = tf.Session()
sess.run(tf.global_variables_initializer())
print(sess.run(W1))
print(sess.run(B1))
x_train = [[0, 0], [0, 1], [1, 0], [1, 1]]
print(sess.run(Z, feed_dict={X: x_train}))
```

2번째 히든 레이어에서 1번째 히든 레이어의 출력값(Z1)과 바이어스(B2)을 받아서 1개의 시그모이드 함수 결과(Y_hat) 출력한다.

2개의 Z(z1, z2)와 바이어스(bias)를 입력으로 받는 1개의 뉴런을 만든다.

```python
W2 = tf.Variable(tf.random_normal([2, 1], dtype=tf.float32))
B2 = tf.Variable(tf.zeros([1]))
Y_hat = tf.sigmoid(tf.matmul(Z, W2) + B2)
```

```python
sess.run(tf.global_variables_initializer())
print(sess.run(W2))
print(sess.run(B2))
print(sess.run(Y_hat, feed_dict={X: x_train}))
```

손실 함수 크로스 엔트로피를 사용하고 경사 하강법으로 모델의 매개 변수(가중치, 바이어스)를 최적화 한다.

```python
# 크로스 엔트로피
loss = tf.reduce_mean(
    -1 * (Y * tf.log(Y_hat) + (1.0 - Y) * tf.log(1.0 - Y_hat)) 
)

# 경사 하강법
train = tf.train.GradientDescentOptimizer(0.05).minimize(loss)
```

학습 시킨다.

```python
sess = tf.Session()
sess.run(tf.global_variables_initializer())

x_train = [[0, 0], [0, 1], [1, 0], [1, 1]] # 피쳐
y_train = [[0], [1], [1], [0]] # 레이블

for epoch in range(40001):
    sess.run(train, feed_dict={X: x_train, Y: y_train})
    if epoch % 5000 == 0:
        print('epoch: {}'.format(epoch))
        print('output\n', sess.run(Y_hat, feed_dict={X: x_train}), sep='')
```
