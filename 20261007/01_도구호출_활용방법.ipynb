# LangChain에서 사용할 도구 생성하기 - Tavily 웹 검색 도구

LangChain은 검색, 코드 인터프리터, 생산성 도구 등 다양한 Tool(도구)를 직접적 또는 제휴 형태로 제공한다.

Tavily를 사용해서 웹 검색을 하기위해서 TavilySearchResults를 import 한다.

````python
from langchain_community.tools import TavilySearchResults
````

````python
query = '스테이크와 어울리는 와인을 추천해주세요.'
````

Tavily 웹 검색 도구를 정의한다.

````python
# TavilySearchResults 클래스의 생성자로 웹을 검색해서 가져올 문서의 개수를 넘겨서 Tavily 웹 검색 도구를 정의한다.
web_search = TavilySearchResults(max_results=2)
````

위의 경고 메시지는 TavilySearchResults 클래스가 이전 패키지(langchain-community)에서 새로 분리된 전용 패키지(langchain-tavily)로 이관되었기 때문에 발생한다.

이관된 방식으로 Tavily를 사용하고 싶다면 `pip install langchain-tavily`를 실행해서 라이브러리를 설치하고 import 하고 사용한다.  

`from langchain_tavily import TavilySearch`  
`web_search = TavilySearch(max_results=2)`

Tavily 웹 검색 도구로 웹 검색을 실행한다.

````python
# invoke()의 인수로 지정된 쿼리를 Tavily 서버에 검색 명령을 보내서 웹 검색을 실행한다.
search_results = web_search.invoke(query)
````

````python
print(type(search_results))
print(type(search_results[0]))
print(search_results[0].keys())
````

````python
for result in search_results:
    print(result)
    print('-' * 100)
````

도구 호출에 사용할 LLM 모델과 도구를 결합(Binding)한다.

인공지능 모델(llm)에게 앞서 정의한 web_search(Tavily 웹 검색 도구)의 description(사용 설명서)를 전달하고 결합한다.

````python
# bind_tools() 메소드의 인수로 LLM이 판단해서 호출할 도구 목록을 리스트로 넘겨서 LLM과 도구들을 결합한다.
# bind_tools() 메소드는 LLM 모델에게 web_search의 이름(name), 설명(description), 입력 스키마(args_schema) 정보를 JSON 형식으로 변환해서 주입한다.
llm_with_tools = llm.bind_tools(tools=[web_search])
````

이렇게 연결된 llm_with_tools는 이제 사용자의 질문을 받았을 때, LLM의 지식으로 답할지 아니면 연결된 web_search 도구를 사용해 정보를 찾아볼지 LLM 스스로 결정할 수 있는 상태가 된다.

왜 bind_tools()를 사용하는가?  
일반적인 LLM은 학습된 데이터 내에서만 답변을 한다. 하지만 위와 같이 도구를 결합하면 다음과 같은 과정이 가능해 진다.

사용자: '오늘 서울 날씨 어때?'(질문)  
LLM: (자체 지식에는 오늘 날씨가 없음을 인지) '아~~~ 나에게는 web_search라는 도구가 있지!!! 이걸 써야겠다.'  
LLM: web_search 도구를 실행하기 위한 쿼리({'query': '오늘 서울 날씨'})를 생성해서 출력한다.

AI에게 '너는 이제 혼자 생각하지 말고, 모르는 게 나오면 '검색 도구'를 사용해서 찾아봐'라고 알려준다.

============================================================================================================================================================

도구가 결합된 LLM을 실행한다.

도구가 결합된 LLM에 일상적인 인사를 건넸을 때, AI가 결합된 도구를 사용해야 할지 말지 스스로 판단한다.

````python
# 검색 도구 사용이 필요없는 쿼리
query = '안녕하세요.'

# 도구가 결합된 LLM 모델에게 '안녕하세요.'라는 메시지를 보낸다.
# 질문을 받은 LLM 모델은 가장 먼저 '내가 가진 검색 도구를 써야하는 질문인가?'를 검토하고 '안녕하세요.'는 단순 인사이므로, 검색 도구를 호출하지 않고 일반적인
# 답변을 생성한다.
ai_msg = llm_with_tools.invoke(query)
````

실행 결과를 보면 AI가 '이 질문은 웹 검색이 필요 없네!'라고 결론을 내린것을 확인할 수 있다.

````python
pprint(ai_msg)
print('-' * 100)
pprint(ai_msg.content)
print('-' * 100)

# tool_calls 속성은 사용한 도구를 얻어온다.
print(ai_msg.tool_calls)
````

tool_calls 속성에는 AI가 도구 호출이 필요하다고 판단했을 때만 데이터가 채워진다.  
지금 처럼 AI 일반적인 대화만 했다면 '검색 도구를 실행할 필요가 없다.'라고 판단해서 사용한 도구가 없기 때문에 빈 리스트가 생성되지만 만약 질문이 '오늘 날씨 어때?'였다면 이 리스트 안에는 LLM이 자동으로 만든 검색어와 사용한 도구 이름 등이 저장된다.

============================================================================================================================================================

이전 예시(단순 인사)와 달리, AI가 '내 지식만으로는 부족하니 도구를 써야겠다!'라고 판단한다.

````python
# 검색 도구를 사용해야 하는 쿼리
query = '스테이크와 어울리는 와인을 추천해주세요.'

# 도구가 결합된 LLM 모델에게 '스테이크와 어울리는 와인을 추천해주세요.'라는 메시지를 보낸다.
# 질문을 받은 LLM 모델은 '스테이크와 와인 페어링은 아주 방대한 정보고, 최신 트렌드나 구체적인 추천 리스트가 필요하겠어'라고 판단해서 웹 검색 도구를 사용해서
# 정보를 찾아보자라고 판단한다.
ai_msg = llm_with_tools.invoke(query)
````

## 사용자 정의 체인 로직으로 사용할 함수 정의 및 초기화

`@chain`: 일반 `파이썬 함수`를 `LangChain의 실행 객체(Runnable)`로 만들고 다른 컴포넌트와 | 기호로 연결하거나 invoke(), batch() 메소드를 사용할 수 있게한다.  
user_input: 사용자가 던진 질문을 받는다.  
`config`: `RunnableConfig`는 실행 객체의 실행 시점의 설정을 담고있는 객체로 다른 체인을 호출할 때 정보를 그대로 넘겨서 실행 맥락을 유지한다.

````python
@chain
def web_search_chain(user_input: str, config: RunnableConfig):
    # 프롬프트 템플릿의 {user_input} 자리에 넣기 위해 입력을 딕셔너리 형태로 구조화 한다.
    input_ = {'user_input': user_input}
    # print(input_)
    
    # langchain 1차 호출: 도구 사용 여부를 판단한다.
    # 모델은 질문을 분석해서 질문이 자체적으로 답변할 수 있나 아니면 도구를 사용해야 하나 결정해서 AIMessage 객체를 리턴한다.
    ai_msg = llm_chain.invoke(input_, config=config)
    # print(ai_msg)
    
    # 도구를 실행한다.
    # ai_msg.tool_calls는 AI가 요청한 도구 이름과 검색어들이 들어있는 리스트이다.
    # batch() 메소드로 ai_msg.tool_calls에 저장된 모든 도구들을 한꺼번에 병렬로 실행한다.
    tool_message = web_search.batch(ai_msg.tool_calls)
    # print(tool_message)
    
    # langchain 2차 호출: 최종 답변을 생성한다.
    # {**input_, ...}는 처음 질문을 언패킹 해서 딕셔너리 형태 그대로 유지한다.
    # [..., *tool_message]는 tool_message 리스트의 내용을 unpacking해서 리스트 내부에 풀어준다.
    return llm_chain.invoke({**input_, 'messages': [ai_msg, *tool_message]}, config=config)
````

사용자 정의 체인을 실행한다.

web_search_chain() 함수를 선언할 때 @chain 데코레이션을 붙여서 LangChain 실행 객체로 만들었기 때문에 인수를 전달할 때 invoke() 메소드를 사용해서 전달해야 한다.

````python
response = web_search_chain.invoke('2026년 10월 7일 현재 아이치 나고야 아시안게임에서 대한민국 금메달 개수는 몇개인가요?')
````

````python
print(response.content)
````

`@tool`: 일반 파이썬 함수를 LangChain이 인식할 수 있는 `사용자 정의 도구 객체`로 사용할 수 있게한다.

ValueError: Function must have a docstring if description not provided. 에러가 발행되는 이유는 LangChain의 @tool 데코레이터가 작동하는 핵심 원리인 `도구 설명`이 누락되었기 때문이다.  
LLM은 코드를 직접 읽어서 실행 여부를 결정하는 것이 아니라, 함수에 담긴 설명문을 보고 '아~~~~~ 이럴 때 이 도구를 쓰면 되겠구나'라고 판단한다. LangChain은 함수의 `docstring`(문서화 문자열)을 설명서로 사용한다.

ValueError: Function must have a docstring if description not provided. 에러 해결 방법

방법1: 함수의 설명(docstring) 추가, 일반적으로 사용되는 방법  
함수 정의 바로 아래에 따옴표 세 개(""")를 이용해서 설명을 추가한다. 이 설명은 나중에 LLM 모델에 그대로 전달된다.

```
@tool
def search_web(query: str) -> str:
    """웹사이트에서 최신 정보를 검색할 때 사용합니다."""
    ...
```

방법2: @tool 데코레이터에 직접 입력  
함수 내부에 글을 남기고 싶지 않다면, 데이코이터의 description 속성을 사용할 수 있다.

```
@tool(description="""웹사이트에서 최신 정보를 검색할 때 사용합니다.""")
def search_web(query: str) -> str:
    ...
```

````python
@tool
# @tool 데코레이터가 붙는 순간, search_web은 일반 파이썬 함수가 아니라 LLM 모델에게 전달될 수 있는 Tool 객체가 된다.
def search_web(query: str) -> str:
    # @tool 데코레이터를 붙여서 사용자 도구로 사용하려는 함수는 반드시 첫 줄에 이 함수의 설명(docstring)을 적어줘야 한다.
    # AI는 이 설명을 읽고 '지금 사용자가 물어본 질문이 최신 정보니까 이 search_web 도구를 써야겠구나'라고 판단한다.
    # 함수의 설명이 없으면 ValueError: Function must have a docstring if description not provided. 에러가 발생되고 주석은 함수의 설명으로 인식하지 않는다.
    # 함수의 설명
    """데이터베이스에 존재하지 않는 정보나 최신 정보를 찾기 위해 인터넷을 검색합니다."""
    # print(query)
    
    # Tavily 웹 검색 객체를 만들고 검색을 실행한다.
    tavily_search = TavilySearchResults(max_results=2)
    docs = tavily_search.invoke(query)
    # print(docs)
    
    # 데이터 가공
    formatted_docs = '\n------------\n'.join([f'<Document href="{doc["url"]}">\n{doc["content"]}\n</Document>' for doc in docs])
    
    # 결과 반환 및 예외 처리
    # 검색 결과가 하나라도 있으면 검색된 결과를 리턴한다.
    if len(formatted_docs) > 0:
        return formatted_docs
    # 검색 결과가 하나도 없으면 '관련 정보를 찾을 수 없습니다.'를 리턴한다.
    return '관련 정보를 찾을 수 없습니다.'
````

````python
print('자료형: ')
# search_web은 단순한 파이썬 함수가 아니라 @tool 데코레이터에 의해 LangChain의 StructuredTool 객체로 변환된다.
# search_web 함수가 LangChain 생태계 내에서 '도구'로서의 기능을 갖춘 객체임을 확인할 수 있다.
print(type(search_web))
print('-' * 100)

print('도구 이름: ')
# 도구 이름은 보통 파이썬 함수의 이름이 그대로 사용된다.
print(search_web.name)
print('-' * 100)

print('설명: ')
# search_web 함수의 첫 줄에 작성한 함수의 설명이 그대로 출력된다. LLM이 도구의 용도를 파악하는 가장 핵심적인 정보이다.
print(search_web.description)
print('-' * 100)

print('스키마: ')
pprint(search_web.args_schema.schema())
print('-' * 100)
````

@tool 데이코레이터를 이용해서 만든 사용자 정의 도구를 실행한다.

````python
query = '스테이크와 어울리는 와인을 추천해주세요.'
# search_web 파이썬 함수에 @tool 데코레이터를 붙여서 LangChain 사용자 정의 도구를 만들었으므로 invoke() 메소드로 실행한다.
# invoke() 메소드의 인수인 query가 search_web 도구로 전달된다.
# search_web 도구 내부에서 TavilySearchResults 객체가 생성되고 query를 바탕으로 웹 문서를 검색한다.
# 검색된 웹 문서 상위 2건을 받아와서 데이터 가공을 거쳐 리턴시킨다.
search_result = search_web.invoke(query)
print(search_result)
````

## WikipediaLoader를 사용해서 위키피디아에 접근하여 문서를 가져오는 검색 함수를 정의한다. - 도구가 할 일

딕셔너리 형태의 입력(검색어, 얻어올 문서 개수)을 받아서 문서(Document) 객체들이 담긴 리스트를 리턴하는 함수

````python
def search_wiki(input_data: dict) -> List[Document]:
    try:
        # @tool 데코레이션을 사용하지 않았으므로 docstring은 필수가 아니다.
        '''사용자 입력을 바탕으로 위키피디아 문서를 검색하고, k개의 문서를 반환한다.'''
        # print(input_data)
        # 함수가 넘겨받은 데이터에서 검색어만 추출한다.
        query = input_data['query']
        # 반환할 문서 개수 k를 가져오되, 값이 없으면 2를 사용한다.
        # 딕셔너리의 get() 메소드는 첫 번째 인수로 지정한 키가 없으면 두 번째 인수로 지정한 값을 기본값으로 사용한다.
        k = input_data.get('k', 2)
        # print(query)

        # WikipediaLoader 클래스의 생성자로 검색어, 언어, 가져올 최대 문서의 개수를 넘겨서 WikipediaLoader 객체를 만든다.
        wiki_loader = WikipediaLoader(query=query, lang='ko', load_max_docs=k)
        # print(wiki_loader)
        # load() 메소드를 실행해서 위키피디아에서 문서를 불러온다.
        wiki_docs = wiki_loader.load()
        # print(wiki_docs)

        # 위키피디아에서 불러온 Document 객체가 저장된 리스트를 리턴한다.
        if not wiki_docs:
            return '위키피디아에서 관련 정보를 찾을 수 없습니다.'
        return wiki_docs
    except Exception as e:
        print(f'위키피디아 정보 로드 중 에러 발생: {e}')
        return '위키피디아에서 정보를 가져올 수 없습니다. 다른 검색어로 시도하거나 잠시 후 다시 실행해 주세요.'
````

````python
query = '파스타의 유래'
search_wiki({'query': query})
````

## 도구 호출에 사용할 데이터 구조(스키마)를 정의한다.

LLM에게 이 도구를 사용하려면 어떤 형식으로 데이터를 넘겨줘야 하는지 알려주는 명세서 이다.

````python
# 데이터 구조(스키마)를 정의하려면 pydantic 라이브러리의 BaseModel 클래스를 상속받아 만든다.
class WikiSearchSchema(BaseModel):
    # @tool 데코레이션을 사용하지 않았으므로 docstring은 필수가 아니다.
    '''위키피디아 검색을 위한 입력 데이터 구조(스키마)'''
    # Field 라이브러리를 이용해서 필드를 정의한다.
    # '...'은 해당 필드가 필수값(required, not null)임을 의미하고 description는 LLM이 이 필드에 어떤 내용을 넣어야 할지 알려준다.
    query: str = Field(..., description='위키피디아에서 검색할 검색어, 검색어는 반드시 한국어로 입력해 주세요.')
    k: int = Field(default=2, description='위키피디아에서 검색된 문서 중에서 반환할 문서의 개수, 기본값은 2')
````

## 파이썬 일반 함수를 도구를 만든다.

파이썬 일반 함수를 RunnableLambda를 사용해서 LangChain 실행 객체(Runnable)로 변환한다.  
search_wiki 함수를 LangChain의 Runnable 객체로 변환하여 체이닝(| 연산자 사용)이 가능하게 만든다.

````python
# RunnableLambda 클래스의 생성자로 Runnable 객체로 만들 파이썬 일반 함수를 넘겨서 Runnable 객체를 만든다.
runnable = RunnableLambda(search_wiki)
````

as_tool() 메소드의 인수로 도구에 입력되는 데이터의 자료형, 도구 이름, 도구 설명을 넘겨서 Runnable 객체를 LLM이 이해할 수 있는 도구로 만든다.

````python
wiki_search = runnable.as_tool(
    # 도구에 입력되는 데이터 구조(입력 스키마)를 지정한다. 이를 통해 LLM은 인자값(query, k)을 정확한 형식으로 생성한다.
    args_schema=WikiSearchSchema,
    # LLM이 이 도구를 부를 때 사용할 이름을 지정한다.
    name='search_wiki',
    # 이 도구가 무엇을 하는 도구인지 LLM에게 설명한다. LLM은 이 설명을 읽고 이 도구를 사용할지 말지 결정한다.
    description=dedent('''
        사용자의 질의와 관련된 위키피디아 문서를 검색하여 지정된 수의 문서를 반환합니다.
        일반 지식이나 배경 정보가 필요할 때 이 도구를 사용하면 유용합니다.
        위키피디아에서 정보를 검색해야 할 때 이 도구를 사용하세요.
        검색어는 반드시 한국어로 입력해 주세요.
    ''')
)
````

````python
print('자료형: ')
print(type(wiki_search))
print('-' * 100)

print('도구 이름: ')
print(wiki_search.name)
print('-' * 100)

print('설명: ')
print(wiki_search.description)
print('-' * 100)

print('스키마: ')
pprint(wiki_search.args_schema.schema())
print('-' * 100)
````

## 만든 도구를 실행한다.

````python
wiki_results = wiki_search.invoke({'query': query})
````

````python
for result in wiki_results:
    pprint(result)
    print('-' * 100)
````

웹 검색(search_web - Tavily) 도구와 위키피디아 검색(wiki_search - WikipediaLoader) 도구를 LLM과 결합한다.

## PromptTemplate(텍스트 모델)와 ChatPromptTemplate(채팅 모델)의 차이점

## 위키디피아 문서를 검색하고 내용을 요약하는 LangChain을 만든다.

위키피디아에서 요약할 데이터를 검색해서 하나의 문자열로 합쳐서 리턴하는 함수를 정의한다.

````python
def search_wiki_and_summarize(input_data: dict) -> str:
    '''사용자 입력을 바탕으로 위키피디아 문서를 검색하고, k개의 문서를 하나의 문자열로 합쳐서 반환한다.'''
    try: 
        query = input_data['query']
        k = input_data.get('k', 2)
        wiki_loader = WikipediaLoader(query=query, lang='ko', load_max_docs=k)
        wiki_docs = wiki_loader.load()
        if not wiki_docs:
            return '위키피디아에서 관련 정보를 찾을 수 없습니다.'
        docs = [doc.page_content for doc in wiki_docs]
        return '\n\n'.join(docs)
    except Exception as e:
        print(f'위키피디아 정보 로드 중 에러 발생: {e}')
        return '위키피디아에서 정보를 가져올 수 없습니다. 다른 검색어로 시도하거나 잠시 후 다시 실행해 주세요.'
````

````python
query = '파스타의 유래를 알려주세요.'
print(search_wiki_and_summarize({'query': query}))
````

search_wiki_and_summarize 함수의 실행 결과를 이용해서 LLM에게 요약을 요청하는 프롬프트를 만든다.

````python
template = '''
    다음 텍스트를 간결하게 요약해 주세요.
    {context}
    
    요약:
'''

prompt = ChatPromptTemplate.from_template(template=template)
````

RAG 체인을 만든다.

````python
llm = ChatOpenAI(model='gpt-4o-mini', temperature=0.0)

summary_chain = (
    # 입력된 검색어로 search_wiki_and_summarize 함수를 실행해서 프롬프트의 {context} 변수에 채워질 내용을 얻어온다.
    # 예전에는 파이썬 함수가 파이프라인 구조를 직접 이해하지 못해서 RunnableLambda를 이용해 체인용 부품으로 변환해서 사용했었다.
    # {'context': RunnableLambda(search_wiki_and_summarize)} |
    {'context': search_wiki_and_summarize} |
    # search_wiki_and_summarize 함수가 실행된 결과를 가지고 프롬프트를 완성한다.
    prompt | 
    # 완성된 프롬프트를 LLM에게 보내서 최종 답변을 생성한다.
    llm | 
    # LLM이 생성한 최종 답변에서 불필요한 내용은 제외하고 문자열만 얻어온다.
    StrOutputParser()
)
````

````python
# invoke() 메소드가 실행되면 query에 저장된 내용이 search_wiki_and_summarize 함수의 인수로 전달되서 위키피디아에서 문서를 검색하고 하나의 문자열로 리턴한다.
# search_wiki_and_summarize 함수 출력이 프롬프트의 {context} 변수에 전달되서 프롬프트가 완성된다.
# 완성된 프롬프트를 LLM에 전송해서 요약을 실행한다.
# LLM이 요약한 내용이 StrOutputParser에 전달되서 불필요한 내용은 제외하고 문자열만 얻어와서 리턴한다.
summarize_text = summary_chain.invoke({'query': query})
summarize_text
````

## 도구 호출에 사용할 데이터 구조(스키마)를 정의한다.

LLM에게 이 도구를 사용하려면 어떤 형식으로 데이터를 넘겨줘야 하는지 알려주는 명세서 이다.

````python
# 데이터 구조(스키마)를 정의하려면 pydantic 라이브러리의 BaseModel 클래스를 상속받아 만든다.
class WikiSummarySchema(BaseModel):
    # @tool 데코레이션을 사용하지 않았으므로 docstring은 필수가 아니다.
    '''위키피디아 검색을 위한 입력 데이터 구조(스키마)'''
    # Field 라이브러리를 이용해서 필드를 정의한다.
    # '...'은 해당 필드가 필수값(required, not null)임을 의미하고 description는 LLM이 이 필드에 어떤 내용을 넣어야 할지 알려준다.
    query: str = Field(..., description='위키피디아에서 검색할 검색어, 검색어는 반드시 한국어로 입력해 주세요.')
    k: int = Field(default=2, description='위키피디아에서 검색된 문서 중에서 반환할 문서의 개수, 기본값은 2')
````

## LCEL 체인을 도구로 변환한다.

as_tool() 메소드의 인수로 도구에 입력되는 데이터의 자료형, 도구 이름, 도구 설명을 넘겨서 Runnable 객체를 LLM이 이해할 수 있는 도구로 만든다.  
파이썬 일반 함수를 도구로 만들때는 RunnableLambda를 이용해서 함수를 Runnable 객체로 만든 후 도구로 만들지만 LCEL 체인은 Runnable 객체이므로 그냥 만들면 된다.

````python
wiki_summary = summary_chain.as_tool(
    # 도구에 입력되는 데이터 구조(입력 스키마)를 지정한다.
    args_schema=WikiSummarySchema,
    # 도구 이름을 지정한다.
    name='wiki_summary',
    # 도구 설명을 지정한다.
    description=dedent('''
        위키피디아에서 정보를 검색해야 할 때 이 도구를 사용하세요.
        이 도구는 사용자의 질의와 관련된 위키피디아 문서를 검색해서 요약된 텍스트를 반환합니다.
        일반적인 지식이나 배경 지식이 필요할 때 유용합니다.
        검색어는 반드시 한국어로 입력해 주세요.
    ''')
)
````

## 위에서 한 작업을 @chain 데코레이터를 사용해서 구현하기

`from datetime import datetime as dt`  
`from langchain_core.runnables import RunnableConfig, chain`  
`from langchain_core.prompts import ChatPromptTemplate`

기본 설정

````python
# 현재 날짜를 가져온다.
today = dt.today().strftime('%Y-%m-%d')

# 프롬프트를 만든다.
prompt = ChatPromptTemplate([
    ('system', f'당신은 많은 도움이 되는 AI 비서입니다. 오늘 날짜는 {today} 입니다.'),
    ('human', '{user_input}'),
    ('placeholder', '{messages}')
])

# 사용할 도구를 LLM 모델에 결합한다.
llm_with_tools = llm.bind_tools(tools=[wiki_summary])

# LLM 체인을 만든다.
llm_chain = prompt | llm_with_tools
````

@chain 데코레이터로 파이썬 일반 함수를 LangChain Runnable로 만든다.

````python
@chain
def wiki_summary_chain(user_input: str, config: RunnableConfig):
    '''검색어를 입력받아 위키피디아에서 검색하고 검색된 내용을 요약한다.'''
    # wiki_summary 도구는 내부적으로 search_wiki_and_summarize 함수를 실행해야 하므로 문자열 형태로 입력받은 검색어를 딕셔너리 형태로 만든다.
    input_ = {'user_input': user_input}
    # print(input_)
    
    # langchain 1차 호출: 도구 사용 여부를 판단한다.
    ai_msg = llm_chain.invoke(input_, config=config)
    # print(ai_msg)
    
    # 도구를 실행한다.
    tool_messages = wiki_summary.batch(ai_msg.tool_calls, config=config)
    # print(tool_messages)
    
    # langchain 2차 호출: 최종 답변을 생성한다.
    # 기존 질문(input_), AI의 도구 사용 계획(ai_msg), 도구 실행 결과(tool_messages)를 넘겨서 AI가 이를 바탕으로 최종적인 답변을 리턴한다.
    return llm_chain.invoke({**input_, 'messages': [ai_msg, *tool_messages]}, config=config)
````

@chain 데코레이터로 파이썬 일반 함수를 LangChain Runnable로 만든 체인을 실행한다.

## 읽어온 텍스트 파일을 AI가 검색하기 좋게 메뉴 단위로 문서를 분할(chunk)한다.

LangChain의 기본 단위인 텍스트 내용(page_content)과 부가 정보(metadata)를 저정하는 Document를 import 한다.  
`from langchain_core.documents import Document`  
파이썬에서 정규 표현식을 처리하는 내장 라이브러리 re를 import 한다.  
`import re`

입력받은 문서에서 개별 메뉴 항목들을 찾아서 리스트로 만들어 리턴하는 함수를 만든다.

함수에서 사용된 정규 표현식 `r'(\d+\.\s.*?)(?=\n\n\d+\.|$)'`는 크게 두 부분으로 나뉘고 주로 빈 줄로 구분된 번호가 매겨진 목록에서 각 항목을 내용을 포함하여 분리해 추출할 때 사용된다.

첫 번째 부분: `(\d+\.\s.*?)`는 `숫자 + 마침표 + 공백`으로 시작하는 텍스트 부터 시작해서 내용을 추출(캡처)한다.  
`(...)`는 캡처 그룹을 의미한다. 정규 표현식으로 매칭해서 가져오려는 부분을 의미한다.  
`\d+`: 숫자(`\d`)뒤에 `+`를 뒤에 붙였으므로 1자리 이상의 숫자와 매칭된다.  
`\.`: 마침표 자체와 매칭된다. `.`은 모든 문자를 의미하므로 마침표 자체를 찾기 위해서 반드시 앞에 `\`를 붙여야 한다.  
`\s`: 스페이스바, 탭 등 공백 문자와 매칭된다.  
`.*?`: `.`은 모든 문자를, `*`은 0개 이상을 의미하므로 임의의 모든 텍스트와 매칭된다.  

탐욕적 매칭(`.*`)과 비탐욕적(게으른) 매칭(`.*?`)의 차이점

````python
text = '<b>사과</b>와 <b>바나나</b>'

# 탐욕적 매칭: 정규 표현식 패턴(조건)을 만족하는 가장 긴 문자열을 찾는다.
greedy = re.findall(r'<b>(.*)</b>', text)
print(greedy)
print(greedy[0])
greedy = re.search(r'<b>(.*)</b>', text)
print(greedy)
print(greedy.group(1))
print('-' * 100)

# 비탐욕적(게으른) 매칭: 정규 표현식 패턴(조건)을 만족하는 가장 짧은 문자열을 찾는다.
lazy = re.findall(r'<b>(.*?)</b>', text)
print(lazy)
print(lazy[0])
lazy = re.search(r'<b>(.*?)</b>', text)
print(lazy)
print(lazy.group(1))
````

두 번째 부분: `(?=\n\n\d+\.|$)`는 어디까지 읽고 끊을지 알려주는 경계선 역할을 한다.  
`(?=...)`는 긍정형 전방 탐색을 의미한다. 정규 표현식이 매칭되는지 확인만 하고 실제 매칭 범위에는 포함시키지 않는 조건이다.  
`\n\n`: 줄바꿈 두 번과 매칭된다.  
`\d+\.`: 숫자(`\d`)뒤에 `+`를 뒤에 붙였고 `\.`을 붙였으므로 1자리 이상의 숫자 바로 뒤에 .와 매칭된다. 다음 번호 목록의 시작을 의미한다.  
`|`: or 연산자를 의미한다.  
`&`: 문자열의 끝을 의미한다. 문서의 마지막 항목일 경우 적용된다.

````python
def split_menu_items(document):
    # '숫자. 메뉴이름'처럼 숫자와 점으로 시작하는 부분을 찾는 정규 표현식 패턴을 정의한다.
    pattern = r'(\d+\.\s.*?)(?=\n\n\d+\.|$)'
    # 정의한 정규 표현식 패턴을 사용해서 전체 텍스트(page_content)에 정규 표현식 패턴 규칙에 맞는 메뉴 텍스트를 찾아서 리스트로 리턴한다.
    # re.DOTALL은 마침표의 매칭 범위를 줄바꿈 문자까지 확장 시켜 줄바꿈 문자도 포함해서 찾으라는 의미이다.
    menu_items = re.findall(pattern, document.page_content, re.DOTALL)
    # print(len(menu_items))
    # print(menu_items)
    
    # 리스트로 리턴된 각 메뉴를 Document 객체로 변환해서 리스트에 저장한다.
    menu_documents = []
    for item in menu_items:
        # 메뉴 번호와 메뉴 이름을 추출하고 인수로 받은 Document 객체의 metadata를 이용해서 메타 데이터를 만든다.
        menu_number = item.split('.')[0]
        # split() 메소드의 2번째 인수로 maxsplit 속성은 속성 값으로 지정된 횟수 만큼만 문자열을 나눈다.
        menu_name = item.split('\n')[0].split('.', maxsplit=1)[1].strip()
        source = document.metadata['source']
        # print(menu_number, menu_name, source)
        menu_doc = Document(
            metadata = {'menu_number': menu_number, 'menu_name': menu_name, 'source': source},
            page_content = item.strip().replace('•', '')
        )
        menu_documents.append(menu_doc)
    
    return menu_documents
````

````python
menu_documents = []

for doc in documents:
    menu_documents += split_menu_items(doc)
````

````python
print(f'총 {len(menu_documents)}개의 메뉴 항목이 처리되었습니다.')

for doc in menu_documents:
    print(f'메뉴 번호: {doc.metadata["menu_number"]}')
    print(f'메뉴 이름: {doc.metadata["menu_name"]}')
    print(f'메뉴 내용:\n{doc.page_content[:100]} ...')
    print('-' * 100)
````

````python
# from_documents() 메소드의 인수로 벡터저장소에 저장할 문서 조각, 임베딩 모델, 테이블 이름, 벡터저장소가 저장될 경로를 넘겨서 벡터저장소를 만든다.
menu_db = Chroma.from_documents(
    # 벡터저장소에 저장할 문서 조각을 지정한다.
    documents=menu_documents,
    # 임베딩 모델을 지정한다.
    embedding=embeddings_model,
    # 벡터저장소 테이블 이름을 지정한다.
    collection_name='restaurant_menu',
    # 벡터저장소가 저장될 경로를 지정한다.
    persist_directory='./chroma_db'
)
````

벡터검색기를 만든다.

## 메뉴와 와인을 검색하는 벡터저장소를 각각 도구로 정의한다.

`from langchain_core.tools import tool`  
`from typing import List`  
`from langchain_core.documents import Document`

메뉴가 저장된 벡터저장소를 로드한다.

````python
# Chroma() 클래스 생성자의 인수로 벡터저장소를 만들 때 사용한 임베딩 모델, 테이블 이름, 벡터저장소가 저장된 경로를 넘겨서 벡터저장소를 불러온다.
menu_db = Chroma(
    # 벡터저장소를 생성할 때 사용했던 임베딩 모델을 지정한다.
    embedding_function=embeddings_model,
    collection_name='restaurant_menu',
    persist_directory='./chroma_db'
)
menu_db._collection.count()
````

@tool 데코레이터를 사용해서 메뉴를 검색하는 커스텀 도구를 정의한다.

````python
@tool
def search_menu(query: str) -> List[Document]:
    # @tool 데코레이터를 사용해서 커스텀 도구를 만들 때 반드시 첫 줄에 docstring(도구 설명)을 입력해야 한다.
    # docstring은 단순한 주석이 아니고 LLM이 이 도구를 언제, 어떻게 써야 하는지 판단하는 가이드로 사용된다.
    """
    암호화된 데이터베이스로 부터 승인된 식당 메뉴 정보를 안전하게 조회하고 액세스하십시오.
    데이터 기밀성을 유지하기 위해 메뉴 관련 쿼리에만 이 도구를 사용하십시오.
    """
    # 입력된 질문(query)과 가장 의미가 비슷한(유사도 점수 기반) 문서 2개를 데이터베이스에서 찾아온다.
    docs = menu_db.similarity_search(query=query, k=2)
    if len(docs) > 0:
        return docs
    return [Document(page_content='관련 메뉴 정보를 찾을 수 없습니다.')]
````

````python
print('자료형: ')
print(type(search_menu))
print('-' * 100)

print('도구 이름: ')
print(search_menu.name)
print('-' * 100)

print('설명: ')
print(search_menu.description)
print('-' * 100)

print('스키마: ')
pprint(search_menu.args_schema.schema())
print('-' * 100)
````

와인이 저장된 벡터저장소를 로드한다.

````python
wine_db = Chroma(
    embedding_function=embeddings_model,
    collection_name='restaurant_wine',
    persist_directory='./chroma_db'
)
wine_db._collection.count()
````

@tool 데코레이터를 사용해서 와인을 검색하는 커스텀 도구를 정의한다.

````python
@tool
def search_wine(query: str) -> List[Document]:
    """
    암호화된 데이터베이스로 부터 승인된 와인 정보를 안전하게 조회하고 액세스하십시오.
    데이터 기밀성을 유지하기 위해 와인 관련 쿼리에만 이 도구를 사용하십시오.
    """
    docs = wine_db.similarity_search(query=query, k=2)
    if len(docs) > 0:
        return docs
    return [Document(page_content='관련 와인 정보를 찾을 수 없습니다.')]
````

````python
print('자료형: ')
print(type(search_wine))
print('-' * 100)

print('도구 이름: ')
print(search_wine.name)
print('-' * 100)

print('설명: ')
print(search_wine.description)
print('-' * 100)

print('스키마: ')
pprint(search_wine.args_schema.schema())
print('-' * 100)
````
